## Giới thiệu công cụ

### Sweetviz

Sweetviz là một thư viện Python được thiết kế để phân tích dữ liệu khám phá (Exploratory Data Analysis – EDA) một cách tự động và trực quan.
Công cụ này có khả năng:

* Tự động tổng hợp, mô tả và so sánh các đặc trưng (features) trong tập dữ liệu.

* Sinh ra báo cáo HTML trực quan, trong đó thể hiện:

* Phân phối dữ liệu (Histogram cho biến số, Bar chart cho biến phân loại).

* Thống kê cơ bản: trung bình, trung vị, min, max, độ lệch chuẩn.

* Tỷ lệ dữ liệu thiếu (missing values) và tương quan giữa các biến.


### AutoViz

AutoViz là thư viện Python mạnh mẽ dùng để tự động hóa quá trình trực quan hóa dữ liệu đa biến (Multivariate Analysis).
Khác với Sweetviz chỉ tập trung mô tả, AutoViz giúp tìm mối quan hệ giữa các biến, phát hiện tương quan và xu hướng tiềm ẩn trong dữ liệu.

Công cụ này:

* Tự động xác định loại biến (số, phân loại, nhị phân, ngày tháng).

* Tạo ra biểu đồ scatter plot, boxplot, heatmap, bar chart cho các cặp biến quan trọng.

* Tự động chọn biến “phụ thuộc” (depVar) để xem mối quan hệ với các biến còn lại.

* Giúp người dùng phát hiện các cụm khách hàng, hành vi chi tiêu, và mối tương quan trong dữ liệu.

Ưu điểm:

* Phân tích đa biến (bivariate/multivariate) tự động.

* Trực quan hóa mối quan hệ giữa các thuộc tính.

* Hỗ trợ nhiều loại dữ liệu, kể cả dữ liệu lớn.

## Làm sạch dữ liệu

Một số cột không cần thiết cho việc phân tích nhóm khách hàng và nên được loại bỏ để giảm nhiễu:

* ID: Mã định danh duy nhất của khách hàng. Nó không mang ý nghĩa phân tích và chỉ dùng để nhận dạng khách hàng.

* Dt_Customer: Ngày khách hàng đăng ký. Mặc dù có thể dùng để tính tuổi đời khách hàng, nhưng Recency (thời gian mua hàng gần nhất) thường mang ý nghĩa thực tế hơn cho việc phân tích hành vi.



```python
cols_to_use = ['Year_Birth', 'Education', 'Marital_Status', 'Income', 'Kidhome', 'Teenhome',
 'Recency', 'Complain',
 'MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds',
 'NumDealsPurchases', 'AcceptedCmp1', 'AcceptedCmp2', 'AcceptedCmp3', 'AcceptedCmp4', 'AcceptedCmp5', 'Response',
 'NumWebPurchases', 'NumCatalogPurchases', 'NumStorePurchases', 'NumWebVisitsMonth']
```

### Xử lí dữ liệu trùng lập

```python
# Kiểm tra dữ liệu trùng lặp
duplicate_rows = df[cols_to_use].duplicated()
print('Số hàng trùng lặp:', duplicate_rows.sum())
```

```python
# xóa dữ liệu trùng lặp
marketing_data= df[cols_to_use].drop_duplicates()

# Kiểm tra số hàng sau khi xóa
print('Số hàng sau khi xóa:', len(marketing_data))
```

### Thay thế dữ liệu

```python
#Kiểm tra các giá trị trong cột Kidhome, Teenhome:
print('Kidhome:',marketing_data['Kidhome'].unique())
print('Teenhome:',marketing_data['Teenhome'].unique())
```

```python
# Thay thế dữ liệu của Kidhome, Teenhome bằng 'has teen','has no teen'
marketing_data['Teenhome'] = marketing_data['Teenhome'].replace([0, 1, 2], ['has no teen', 'has teen', 'has teen'])
marketing_data['Kidhome'] = marketing_data['Kidhome'].replace([0, 1, 2], ['has no kid', 'has kid', 'has kid'])

# Kiểm tra sau khi thay đổi:
print('Kidhome:',marketing_data['Kidhome'].unique())
print('Teenhome:',marketing_data['Teenhome'].unique())
```

Nhận xét:


*   Vì Kidhome, Teenhome chứa số lượng trẻ em trong gia đình, nên ta tiến hành thay thế dữ liệu sang 'has no kid' với các giá trị bằng 0, 'has kid' với giá trị lớn hơn 0



### Xử lí dữ liệu bị thiếu

#### Xử lí dữ liệu bị thiếu

```python
# Kiểm tra số lượng dữ liệu bị thiếu
marketing_data.isnull().sum()
```

```python
# Xóa dữ liệu bị thiếu
marketing_data.dropna(subset=['Income'], inplace=True)
print("\n kiểm tra lại:")
print(marketing_data.isnull().sum())
```

### Tạo thêm cột mới

* Để chuẩn bị dữ liệu cho việc phân tích chuyên sâu hơn và xây dựng các mô hình phân khúc khách hàng tiềm năng, chúng ta cần tạo ra các biến tổng hợp quan trọng. Việc này sẽ giúp tóm tắt hành vi chi tiêu và tình trạng gia đình của khách hàng, từ đó tạo ra những hiểu biết sâu sắc hơn về khách hàng.

* Chúng ta sẽ thực hiện hai bước sau:
    1. Tính tổng chi tiêu toàn bộ (Total_Spent): Nhằm đánh giá tổng quan khả năng chi tiêu của mỗi khách hàng trên tất cả các danh mục sản phẩm.

    2. Phân loại tình trạng con cái (Family_Type): Nhằm đơn giản hóa cấu trúc gia đình, phân loại khách hàng thành hai nhóm rõ ràng là 'Có con' và 'Không có con' để xem yếu tố con cái ảnh hưởng như thế nào đến hành vi mua sắm của họ.



```python
# Tổng số tiền chi tiêu
marketing_data['Total_Spent'] = marketing_data[['MntWines','MntFruits','MntMeatProducts',
                        'MntFishProducts','MntSweetProducts','MntGoldProds']].sum(axis=1)

# Kiểm tra xem có con hay không
def family_type(row):
    if row['Kidhome'] == 'has no kid' and row['Teenhome'] == 'has no teen':
        return 'No Children'
    else:
        return 'Has Children'

marketing_data['Family_Type'] = marketing_data.apply(family_type, axis=1)

# Kiểm tra dữ liệu
marketing_data.head()
```

# Kết luận

Qua quá trình phân tích dữ liệu bằng Sweetviz và AutoViz, có thể rút ra một số đặc điểm nổi bật giúp định hình các nhóm khách hàng (customer segments) như sau:

1. Nhóm thu nhập cao (Income > 100.000):

Thường chi tiêu mạnh cho các sản phẩm có giá trị cao như rượu vang (MntWines), thịt (MntMeatProducts) và vàng (MntGoldProds).

Ưa chuộng các kênh mua hàng hiện đại như Catalog và Website, ít quan tâm đến khuyến mãi hoặc giảm giá.
→ Đây là nhóm khách hàng cao cấp - trung thành, có thể tập trung các chiến dịch quảng bá sản phẩm cao cấp.

2. Nhóm thu nhập trung bình (≈ 40.000 – 100.000):

Có hành vi chi tiêu ổn định, chủ yếu mua hàng trực tiếp tại cửa hàng (Store).

Có thể sử dụng một vài ưu đãi giảm giá, nhưng không thường xuyên.
→ Đây là nhóm khách hàng truyền thống, nên ưu tiên chương trình khuyến mãi hoặc thẻ thành viên để tăng tần suất mua hàng.

3.  Nhóm thu nhập thấp (< 40.000):

Nhạy cảm với giá, thường mua khi có giảm giá hoặc ưu đãi trực tuyến.

Chi tiêu tổng thể thấp, ít mua các sản phẩm cao cấp.
→ Là nhóm khách hàng nhạy cảm về giá, phù hợp với các chiến dịch giảm giá ngắn hạn.

4.  Đặc điểm chung về nhân khẩu học:

Đa số khách hàng thuộc nhóm Married hoặc Together, thể hiện sự ổn định trong gia đình.

Phần lớn có trình độ Graduation trở lên, cho thấy tệp khách hàng có mức học vấn và thu nhập tương đối cao.
