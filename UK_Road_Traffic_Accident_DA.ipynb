## **1. Định nghĩa vấn đề (Define Problem)**

### **1.1. Mô tả (Description)**
+ Phạm vi: Dữ liệu về An toàn Giao thông Đường bộ Vương quốc Anh (UK Road Safety: Traffic Accidents and Vehicles) giai đoạn 2005-2017. Dữ liệu được cung cấp chính thức từ UK Department for Transport (Bộ Giao thông Vận tải Anh).
+ Kích thước: Gồm 2 tệp dữ liệu chính với hơn 2 triệu bản ghi về tai nạn và thông tin phương tiện liên quan.
+ Đặc điểm: Gồm 2 tệp (Accident_information và Vehicle_information) liên kết với nhau qua khóa Accident_Index. Dữ liệu có tính uy tín cao, được sử dụng rộng rãi trên Kaggle (140K lượt xem, 17.3K lượt tải).

### **1.2. Dữ liệu vào (Input Data)**
+ **A. Tệp Accident_Information (Thông tin tai nạn):**
  + **Nhóm 1: Thông tin thời gian:**
    + Date: Ngày xảy ra tai nạn.
    + Time: Thời gian xảy ra tai nạn.
    + Day_of_Week: Thứ trong tuần.
    + Year: Năm xảy ra tai nạn.
  + **Nhóm 2: Thông tin địa lí và vị trí:**
    + Latitude: Vĩ độ.
    + Longitude: Kinh độ.
    + Location_Easting_OSGR: Tọa độ đông OSGR.
    + Location_Northing_OSGR: Tọa độ bắc OSGR.
    + Urban_or_Rural_Area: Khu vực thành thị hay nông thôn.
    + Local_Authority_(District): Khu vực hành chính địa phương.
    + Local_Authority_(Highway): Cơ quan quản lý đường bộ địa phương.
    + LSOA_of_Accident_Location: Khu vực thống kê của vị trí tai nạn.
  + **Nhóm 3: Đặc điểm đường và giao lộ:**
    + 1st_Road_Class, 1st_Road_Number: Hạng và số hiệu đường chính
    + 2nd_Road_Class, 2nd_Road_Number: Hạng và số hiệu đường thứ hai
    + Road_Type: Loại đường
    + Speed_Limit: Tốc độ giới hạn
    + Junction_Control, Junction_Detail: Cách kiểm soát và chi tiết giao lộ
  + **Nhóm 4: Điều kiện môi trường và hiện trường:**
    + Light_Conditions: Điều kiện ánh sáng
    + Weather_Conditions: Điều kiện thời tiết
    + Road_Surface_Conditions: Tình trạng bề mặt đường
    + Special_Conditions_at_Site: Điều kiện đặc biệt tại hiện trường
    + Carriageway_Hazards: Nguy cơ/chướng ngại vật trên đường xe chạy
  + **Nhóm 5: Hậu quả và đáp ứng:**
    + Number_of_Vehicles: Số phương tiện liên quan
    + Number_of_Casualties: Số lượng nạn nhân
    + Did_Police_Officer_Attend_Scene_of_Accident: Cảnh sát có mặt tại hiện trường không?
    + Police_Force: Lực lượng cảnh sát ghi nhận
    + Pedestrian_Crossing-Human_Control: Người điều khiển vạch sang đường
    + Pedestrian_Crossing-Physical_Facilities: Cơ sở vật chất cho người đi bộ
    + InScotland: Tai nạn có xảy ra ở Scotland không

+ **B. Tệp Vehicle_Information (Thông tin phương tiện)**

  + **Nhóm 1: Thông tin phương tiện:**
    + Vehicle_Type: Loại xe
    + make, model: Hãng xe và mẫu xe
    + Age_of_Vehicle: Tuổi của xe
    + Engine_Capacity_.CC.: Dung tích động cơ
    + Propulsion_Code: Loại động cơ
    + Was_Vehicle_Left_Hand_Drive: Có phải xe tay lái nghịch không
    + Year: Năm sản xuất phương tiện
  + **Nhóm 2: Thông tin người lái và hành trình:**
    + Sex_of_Driver: Giới tính của tài xế
    + Age_Band_of_Driver: Nhóm tuổi của tài xế
    + Journey_Purpose_of_Driver: Mục đích chuyến đi
    + Driver_Home_Area_Type: Loại khu vực cư trú của tài xế
    + Driver_IMD_Decile: Mức độ thiếu thốn của tài xế
  + **Nhóm 3: Động lực học va chạm:**
    + Vehicle_Manoeuvre: Thao tác của xe trước khi va chạm
    + X1st_Point_of_Impact: Điểm va chạm đầu tiên
    + Vehicle_Leaving_Carriageway: Xe có rời khỏi phần đường xe chạy không
    + Skidding_and_Overturning: Có trượt bánh và lật xe không
    + Hit_Object_in_Carriageway: Vật thể bị va chạm trên làn đường
    + Hit_Object_off_Carriageway: Vật thể bị va chạm ngoài làn đường
    + Junction_Location: Vị trí tại giao lộ
    + Towing_and_Articulation: Tình trạng kéo theo hoặc đầu kéo
    + Vehicle_Location.Restricted_Lane: Vị trí xe trên làn đường bị hạn chế

### **1.3. Dữ liệu ra (Output Data)**
  + Biến mục tiêu (Output variable): **Accident_Severity**: Mức độ nghiêm trọng của tai nạn
  + Kết quả phân tích:
    + Các phát hiện/Insight về thực trạng, xu hướng và các yếu tố then chốt ảnh hưởng đến tai nạn giao thông.
    + Mô hình dự báo mức độ nghiêm trọng của tai nạn.
    + Các giải pháp và khuyến nghị chính sách.

### **1.4. Mục tiêu nghiên cứu (Research Objective)**
+ Phân tích thực trạng và xu hướng tai nạn giao thông theo thời gian, không gian và các điều kiện vận hành.
+ Xác định các yếu tố then chốt ảnh hưởng đến tần suất và mức độ nghiêm trọng của tai nạn.
+ Xây dựng và đánh giá mô hình dự báo mức độ nghiêm trọng tai nạn dựa trên các thuật toán học máy.
+ Đề xuất các giải pháp và khuyến nghị chính sách dựa trên kết quả phân tích.

#### **(2) Kiểm tra tính toàn vẹn của dữ liệu**
+ Dữ liệu có bị trùng lặp không? Hiển thị dòng bị vi phạm.
+ Dữ liệu có tồn tại giá trị Null không? Hiển thị dòng bị vi phạm.
+ Dữ liệu có tồn tại giá trị NaN không? Hiển thị dòng bị vi phạm.

```python
# Kiểm tra giá trị Null hoặc NaN
has_null = Accident.isnull().sum().any()
null_count = Accident.isnull().sum().sum()
if has_null:
    print(f'- Tổng số giá trị Null: {null_count}')
    null_rows = Accident[Accident.isnull().any(axis=1)]
    print(f'- Tổng số dòng có giá trị Null hoặc NaN: {len(null_rows)}')

# Kiểm tra dữ liệu trùng lặp
n_duplicated = Accident.duplicated().sum()
print(f'- Số dòng trùng lặp: {n_duplicated}')

# Thống kê chi tiết các cột có giá trị missing
print("- Thống kê chi tiết giá trị missing theo cột:")
missing_stats = Accident.isnull().sum()
missing_percent = (Accident.isnull().sum() / len(Accident)) * 100
missing_df = pd.DataFrame({
    'Số lượng missing': missing_stats,
    'Tỷ lệ %': missing_percent
}).sort_values('Số lượng missing', ascending=False)
missing_df = missing_df[missing_df['Số lượng missing'] > 0]
display(missing_df)
```

**Nhận xét:**
+ Dữ liệu không có dòng nào bị trùng lặp (0 dòng trùng)
+ Dữ liệu có tồn tại giá trị missing với tổng số 5,017,445 giá trị Null/NaN
+ Có 2,045,811 dòng chứa ít nhất 1 giá trị missing (chiếm ~99.9% tổng số dòng)
+ Các cột Carriageway_Hazards và Special_Conditions_at_Site có tỷ lệ missing rất cao (>97%), có thể cần xem xét loại bỏ hoặc xử lý đặc biệt
+ Các cột 2nd_Road_Class có tỷ lệ missing khá cao (41.24%)
+ Hầu hết các cột khác có tỷ lệ missing thấp (<8%) hoặc rất thấp (<1%)

#### **(3) Các tính chất thống kê trên dữ liệu số**
+ Count, Mean, Standard Deviation, Minimum Value
+ 25th Percentile, 50th Percentile (Median), 75th Percentile, Maximum Value

```python
# Lọc các cột số
numeric_accident = Accident.select_dtypes(include=['number'])
description_accident = numeric_accident.describe().T

# Thêm variance (phương sai)
description_accident['variance'] = numeric_accident.var()
description_accident['datatype'] = numeric_accident.dtypes

# Hiển thị kết quả
display(description_accident[['count', 'mean', 'std', 'variance', 'min', '25%', '50%', '75%', 'max', 'datatype']])
```

#### **(4) Các tính chất thống kê trên dữ liệu phân loại**
+ Count, Unique, Top, Frequency, Data Type.

```python
# Lọc các cột object
object_accident = Accident.select_dtypes(include=['object'])
description_object = object_accident.describe().T

# Thêm các thông tin bổ sung
description_object['unique_count'] = object_accident.nunique()
description_object['most_frequent_count'] = object_accident.apply(lambda x: x.value_counts().iloc[0])
description_object['most_frequent_percent'] = (description_object['most_frequent_count'] / len(Accident)) * 100
description_object['datatype'] = object_accident.dtypes

# Hiển thị kết quả
display(description_object[['count', 'unique', 'top', 'freq', 'most_frequent_count', 'most_frequent_percent', 'datatype']])
```
