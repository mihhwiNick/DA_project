### **1.2. Dữ liệu vào (Input Data)**
+ **A. Tệp Accident_Information (Thông tin tai nạn):**
  + **Nhóm 1: Thông tin thời gian:**
    + Date: Ngày xảy ra tai nạn.
    + Time: Thời gian xảy ra tai nạn.
    + Day_of_Week: Thứ trong tuần.
    + Year: Năm xảy ra tai nạn.
  + **Nhóm 2: Thông tin địa lí và vị trí:**
    + Latitude: Vĩ độ.
    + Longitude: Kinh độ.
    + Location_Easting_OSGR: Tọa độ đông OSGR.
    + Location_Northing_OSGR: Tọa độ bắc OSGR.
    + Urban_or_Rural_Area: Khu vực thành thị hay nông thôn.
    + Local_Authority_(District): Khu vực hành chính địa phương.
    + Local_Authority_(Highway): Cơ quan quản lý đường bộ địa phương.
    + LSOA_of_Accident_Location: Khu vực thống kê của vị trí tai nạn.
  + **Nhóm 3: Đặc điểm đường và giao lộ:**
    + 1st_Road_Class, 1st_Road_Number: Hạng và số hiệu đường chính
    + 2nd_Road_Class, 2nd_Road_Number: Hạng và số hiệu đường thứ hai
    + Road_Type: Loại đường
    + Speed_Limit: Tốc độ giới hạn
    + Junction_Control, Junction_Detail: Cách kiểm soát và chi tiết giao lộ
  + **Nhóm 4: Điều kiện môi trường và hiện trường:**
    + Light_Conditions: Điều kiện ánh sáng
    + Weather_Conditions: Điều kiện thời tiết
    + Road_Surface_Conditions: Tình trạng bề mặt đường
    + Special_Conditions_at_Site: Điều kiện đặc biệt tại hiện trường
    + Carriageway_Hazards: Nguy cơ/chướng ngại vật trên đường xe chạy
  + **Nhóm 5: Hậu quả và đáp ứng:**
    + Number_of_Vehicles: Số phương tiện liên quan
    + Number_of_Casualties: Số lượng nạn nhân
    + Did_Police_Officer_Attend_Scene_of_Accident: Cảnh sát có mặt tại hiện trường không?
    + Police_Force: Vùng Địa lý mà Lực lượng Cảnh sát đó phụ trách quản lý
    + Pedestrian_Crossing-Human_Control: Người điều khiển vạch sang đường
    + Pedestrian_Crossing-Physical_Facilities: Cơ sở vật chất cho người đi bộ
    + InScotland: Tai nạn có xảy ra ở Scotland không

+ **B. Tệp Vehicle_Information (Thông tin phương tiện)**

  + **Nhóm 1: Thông tin phương tiện:**
    + Vehicle_Type: Loại xe
    + make, model: Hãng xe và mẫu xe
    + Age_of_Vehicle: Tuổi của xe
    + Engine_Capacity_.CC.: Dung tích động cơ
    + Propulsion_Code: Loại động cơ
    + Was_Vehicle_Left_Hand_Drive: Có phải xe tay lái nghịch không
    + Year: Năm sản xuất phương tiện
  + **Nhóm 2: Thông tin người lái và hành trình:**
    + Sex_of_Driver: Giới tính của tài xế
    + Age_Band_of_Driver: Nhóm tuổi của tài xế
    + Journey_Purpose_of_Driver: Mục đích chuyến đi
    + Driver_Home_Area_Type: Loại khu vực cư trú của tài xế
    + Driver_IMD_Decile: Mức độ thiếu thốn của tài xế
  + **Nhóm 3: Động lực học va chạm:**
    + Vehicle_Manoeuvre: Thao tác của xe trước khi va chạm
    + X1st_Point_of_Impact: Điểm va chạm đầu tiên
    + Vehicle_Leaving_Carriageway: Xe có rời khỏi phần đường xe chạy không
    + Skidding_and_Overturning: Có trượt bánh và lật xe không
    + Hit_Object_in_Carriageway: Vật thể bị va chạm trên làn đường
    + Hit_Object_off_Carriageway: Vật thể bị va chạm ngoài làn đường
    + Junction_Location: Vị trí tại giao lộ
    + Towing_and_Articulation: Tình trạng kéo theo hoặc đầu kéo
    + Vehicle_Location.Restricted_Lane: Vị trí xe trên làn đường bị hạn chế

###**Bước 3: Hợp nhất dữ liệu**

**- Do mỗi vụ tai nạn có thể liên quan đến nhiều phương tiện khác nhau, bảng Vehicle cần được tổng hợp theo khóa Accident_Index để đảm bảo mỗi bản ghi trong dataset cuối cùng đại diện cho một vụ tai nạn duy nhất, tránh hiện tượng nhân bản bản ghi khi hợp nhất dữ liệu, đồng thời vẫn giữ được thông tin đặc trưng của các phương tiện liên quan đến mỗi vụ tai nạn. Quá trình tổng hợp, sẽ áp dụng theo quy ước:**

  - Đối với các biến phân loại (categorical):
Giá trị đại diện được xác định bằng mode (giá trị xuất hiện nhiều nhất), nhằm phản ánh đặc điểm phổ biến nhất của các phương tiện liên quan trong cùng một vụ tai nạn.
  - Đối với các biến dạng số (numerical):
Sử dụng median (thay vì mean) vì nó không bị ảnh hưởng bởi các giá trị ngoại lai, đồng thời phản ánh xu hướng trung tâm ổn định hơn của dữ liệu.

```python
import polars as pl

# Đổi tên cột để sử dụng
vehicle_clean1 = vehicle_clean1.rename(columns={'Engine_Capacity_.CC.':'Engine_Capacity_CC'})
df = pl.from_pandas(vehicle_clean1)

# Aggregate theo Accident_Index
vehicle_agg_pl = df.group_by("Accident_Index").agg([
    pl.col("Vehicle_Type").mode().first().alias("Vehicle_Type"),
    pl.col("Vehicle_Manoeuvre").mode().first().alias("Vehicle_Manoeuvre"),
    pl.col("Vehicle_Leaving_Carriageway").mode().first().alias("Vehicle_Leaving_Carriageway"),
    pl.col("Age_Band_of_Driver").mode().first().alias("Age_Band_of_Driver"),
    pl.col("Sex_of_Driver").mode().first().alias("Sex_of_Driver"),
    pl.col("Age_of_Vehicle").median().alias("Age_of_Vehicle"),
    pl.col("Engine_Capacity_CC").median().alias("Engine_Capacity_CC")
])

vehicle_agg = vehicle_agg_pl.to_pandas()
print(f"Hoàn thành! Kết quả có {len(vehicle_agg)} dòng")
```

**- Sau khi đã tổng hợp dữ liệu phương tiện theo Accident_Index, chúng ta có bảng vehicle_agg với mỗi dòng đại diện cho một vụ tai nạn duy nhất đồng thời chứa thông tin tổng hợp của các phương tiện liên quan.**

**- Bước tiếp theo, chúng ta sẽ gộp bảng Accident với bảng Vehicle đã tổng hợp để tạo dataset cuối cùng phục vụ cho phân tích khám phá dữ liệu (EDA) và xây dựng mô hình. Với Khóa chính để merge là Accident_Index, sử dụng phương pháp LEFT JOIN để đảm bảo giữ lại tất cả các vụ tai nạn trong accidents_clean kể cả những vụ không có thông tin phương tiện tương ứng.**

```python
merged_data = pd.merge(
    accidents_clean1, # bảng tai nạn
    vehicle_agg,  # bảng vehicle đã aggregate
    on='Accident_Index',
    how='left'
)

merged_data.info()
display(merged_data.head(5))
```

###**Bước 4: Xử lí dữ liệu thiếu (Missing values)**

**- Kiểm tra các giá trị thiếu (Missing Values) của biến số:**

```python
# Lọc các biến số
numerical_cols = merged_data.select_dtypes(include=['float64','int64']).columns

# Thống kê missing chỉ cho biến số
missing_stats = merged_data[numerical_cols].isnull().sum()
missing_percent = (missing_stats / len(merged_data)) * 100

missing_df = pd.DataFrame({
    'Số lượng missing': missing_stats,
    'Tỷ lệ %': missing_percent
}).sort_values('Số lượng missing', ascending=False)

# Chỉ hiển thị những cột có missing
missing_df = missing_df[missing_df['Số lượng missing'] > 0]

display(missing_df)
```

**- Kiểm tra các giá trị thiếu (Missing Values) của biến phân loại:**

```python
# Xác định các biến phân loại
categorical_cols = merged_data.select_dtypes(include=['object']).columns

# Kiểm tra missing values cho biến phân loại
categorical_missing = merged_data[categorical_cols].isnull().sum()
categorical_missing_percent = (categorical_missing / len(merged_data)) * 100
categorical_missing_df = pd.DataFrame({
    'Số lượng missing': categorical_missing,
    'Tỷ lệ %': categorical_missing_percent
}).sort_values('Số lượng missing', ascending=False)
categorical_missing_df = categorical_missing_df[categorical_missing_df['Số lượng missing'] > 0]
display(categorical_missing_df)
```

**Thực hiện xử lí các giá trị thiếu (Missing values) cho cả biến số và biến phân loại theo quy ước:**

- Tỉ lệ missing >50%: xóa cột

- Tỉ lệ mising 5-50%: impute (median) nếu là biến số, impute nhãn (Missing) nếu là biến phân loại.

- Tỉ lệ mising <5%: xóa dòng.

```python
# Xử lý biến số
for col in numerical_cols:
    missing_rate = merged_data[col].isnull().mean()

    if missing_rate == 0:
        continue  # Bỏ qua cột không missing

    if missing_rate > 0.5:
        merged_data.drop(columns=[col], inplace=True)
        print(f"[DROP COL] {col}: missing rate = {missing_rate:.5%} >50%, drop column")

    elif 0.05 <= missing_rate <= 0.5:
        median_val = merged_data[col].median()
        merged_data[col].fillna(median_val, inplace=True)
        print(f"[IMPUTE] {col}: missing rate = {missing_rate:.5%}, fillna with median = {median_val}")

    else:  # missing_rate <5%
        merged_data = merged_data[merged_data[col].notnull()]
        print(f"[DROP ROW] {col}: missing rate = {missing_rate:.5%} <5%, drop rows with NaN")

# Xử lý biến phân loại
for col in categorical_cols:
    missing_rate = merged_data[col].isnull().mean()

    if missing_rate == 0:
        continue  # Bỏ qua cột không missing

    if missing_rate > 0.5:
        merged_data.drop(columns=[col], inplace=True)
        print(f"[DROP COL] {col}: missing rate = {missing_rate:.5%} >50%, drop column")

    elif 0.05 <= missing_rate <= 0.5:
        merged_data[col].fillna('Missing', inplace=True)
        print(f"[IMPUTE] {col}: missing rate = {missing_rate:.5%}, fillna with 'Missing'")

    else:  # missing_rate <5%
        merged_data = merged_data[merged_data[col].notnull()]
        print(f"[DROP ROW] {col}: missing rate = {missing_rate:.5%} <5%, drop rows with NaN")
```

**- Tiếp theo, thực hiện kiểm tra các giá trị không rõ ràng của biến phân loại:**

```python
categorical_cols = merged_data.select_dtypes(include=['object']).columns
total_records = len(merged_data)

special_values = ["Data missing or out of range", "Unknown", "Unallocated", "Other", "Other/Not known", "Not known"]

special_found = False

for col in categorical_cols:
    col_series = merged_data[col].astype(str).str.strip()
    col_series_lower = col_series.str.lower()

    # Đếm từng giá trị đặc biệt
    counts = {}
    for val in special_values:
        if val.lower() in ["unknown", "unallocated"]:
            counts[val] = (col_series_lower == val.lower()).sum()
        else:
            counts[val] = (col_series == val).sum()

    if sum(counts.values()) == 0:
        continue

    if not special_found:
        print(f"{'CỘT':<35} {'GIÁ TRỊ':<40} {'SỐ LƯỢNG':>10}   {'TỶ LỆ':>7}")
        print("-"*100)
        special_found = True

    # In từng giá trị
    for val, count in counts.items():
        if count > 0:
            print(f"{col:<35} {val:<40} {count:>10,}   {count/total_records*100:>6.4f}%")

    print("-"*100)
```

**- Thực hiện gom các giá trị này thành một category chung (Missing). Bao gồm các giá trị: Data missing or out of range, Unknown, Other, Other/Not known, Not known. Sau khi gom, xử lý theo tỉ lệ missing:**

- Nếu tỉ lệ "Missing" >5%: [KEEP] giữ nguyên category "Missing".
- Nếu tỉ lệ "Missing" <5%: [IMPUTE] mode - giá trị xuất hiện nhiều nhất.

```python
missing_labels = [
    "Data missing or out of range",
    "Unknown",
    "Not known",
    "Other",
    "Other/Not known (2005-10)",
    "Unallocated"
]

categorical_cols = merged_data.select_dtypes(include=['object']).columns

for col in categorical_cols:
    # Chuẩn hóa text
    merged_data[col] = merged_data[col].astype(str).str.strip()

    # Gom các nhãn đặc biệt thành 'Missing'
    merged_data[col] = merged_data[col].replace(
        {label: "Missing" for label in missing_labels}
    )

    # Tính tỉ lệ Missing
    missing_rate = (merged_data[col] == "Missing").mean()

    if missing_rate > 0.05:
        # Giữ nguyên category Missing
        print(f"[KEEP] {col}: {missing_rate:.2%} → kept as separate 'Missing' category")
    elif missing_rate > 0:
        # Impute bằng mode
        mode_val = merged_data[col].mode()[0]
        merged_data[col] = merged_data[col].replace("Missing", mode_val)
        print(f"[IMPUTE] {col}: {missing_rate:.2%} → replaced Missing with mode = {mode_val}")
    # Nếu missing_rate == 0, không làm gì
```

**- Dùng biểu đồ Box Plot để trực quan các biến có Outliers:**

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

num_cols = ['Number_of_Casualties', 'Number_of_Vehicles', 'Age_of_Vehicle', 'Engine_Capacity_CC']

# Set style đẹp
sns.set_style("whitegrid")
sns.set_palette("husl")

# Tạo subplot 2x2
fig, axes = plt.subplots(2, 2, figsize=(16, 8))
axes = axes.flatten()

for idx, (ax, col) in enumerate(zip(axes, num_cols)):
    # Tính toán outlier
    Q1 = merged_data[col].quantile(0.25)
    Q3 = merged_data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = merged_data[
        (merged_data[col] < lower_bound) | (merged_data[col] > upper_bound)
    ].shape[0]

    total_count = merged_data.shape[0]
    outlier_percentage = (outlier_count / total_count) * 100

    # Vẽ boxplot ngang
    sns.boxplot(x=merged_data[col],
                ax=ax,
                orient='h',
                width=0.6,
                fliersize=5,
                linewidth=1.5,
                color=f'C{idx}')

    ax.set_title(col, fontsize=14, fontweight='bold', pad=12, color='#2C3E50')
    ax.set_xlabel('Value', fontsize=11)
    ax.set_ylabel('')

    # Thông tin outlier - đặt ở góc trên phải
    ax.text(0.98, 0.95, f'Outlier: {outlier_percentage:.2f}%',
            transform=ax.transAxes, fontsize=11, fontweight='bold',
            verticalalignment='top', horizontalalignment='right',
            bbox=dict(boxstyle='round,pad=0.4', facecolor='white',
                     edgecolor=f'C{idx}', alpha=0.9, linewidth=1.5))

    stats_text = f'Q1: {Q1:.1f} | Q3: {Q3:.1f}\nIQR: {IQR:.1f}'
    ax.text(0.02, 0.95, stats_text,
            transform=ax.transAxes, fontsize=9,
            verticalalignment='top', horizontalalignment='left',
            bbox=dict(boxstyle='round,pad=0.3', facecolor='whitesmoke', alpha=0.7))

    ax.grid(True, linestyle='--', alpha=0.3, axis='x')
    ax.set_axisbelow(True)

    if col == 'Engine_Capacity_CC':
        ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{int(x):,}'))

plt.tight_layout(pad=3.0)
fig.suptitle('Horizontal Boxplot Analysis with Outlier Detection',
             fontsize=16, fontweight='bold', y=1.02)
plt.show()
```

**Nhận xét:**
+ Với biến **Number_of_Casualties(Outlier: 22.21%)** đa số các vụ tai nạn đều có số thương vong rất thấp, thường chỉ 1-2 người. Điều này cho thấy hầu hết là các sự cố nhỏ lẻ. Do đó, giữ lại vì mang thông tin giá trị về các tình huống đặc biệt trong dữ liệu.
+ Với biến **Number_of_Vehicles(Outlier: 2.25%)** cho thấy đa phần là các vụ va chạm đơn giản, chỉ liên quan từ 1-2 phương tiện. Tuy nhiên, sự xuất hiện của các điểm ngoại lai với số xe lớn (67 phương tiện) cho thấy các điểm ngoại lai này phản ánh đúng hiện tượng tai nạn liên hoàn. Do đó, giữ lại vì mang thông tin giá trị về các tình huống đặc biệt trong dữ liệu.

+ Biến **Age_of_Vehicle (Outlier: 9.16%)** phân phối khá tập trung, với độ tuổi trung bình của phương tiện là 7 năm và 75% phương tiện có tuổi đời 9 năm. Tuy nhiên, tồn tại các giá trị ngoại lai lên đến 105 năm - đây rõ ràng là các giá trị bất thường vì không thể có xe cộ lưu thông với tuổi đời cao như vậy. Cần xử lí các giá trị outlier này.
+ Biến **Engine_Capacity_CC.**(Outlier: 12.25%) có phân phối khá rộng với dung tích động cơ trung bình khoảng 1,598 CC. **Thực hiện xử lí các giá trị Outliers bất thường của 2 biến **Age_of_Vehicle** và **Engine_Capacity_.CC. theo quy ước sử dụng IQR Capping:****
+ lower bound = Q1 - 1.5 * IQR
+ upper bound = Q3 + 1.5 * IQR
+ Giá trị < lower bound → gán bằng lower bound
+ Giá trị > upper bound → gán bằng upper bound

```python
num_cols_cap = ['Age_of_Vehicle', 'Engine_Capacity_CC']

summary_list = []

for col in num_cols_cap:
    Q1 = merged_data[col].quantile(0.25)
    Q3 = merged_data[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    print(f"Column: {col}")
    print(f"Lower bound: {lower_bound:.2f}, Upper bound: {upper_bound:.2f}")

    # Capping
    merged_data[col] = np.where(
        merged_data[col] < lower_bound, lower_bound,
        np.where(merged_data[col] > upper_bound, upper_bound, merged_data[col])
    )

    # Lưu summary sau capping để hiển thị trong bảng
    desc = merged_data[col].describe()
    summary_list.append({
        "Variable": col,
        "Count": desc["count"],
        "Mean": desc["mean"],
        "Std": desc["std"],
        "Min": desc["min"],
        "25%": desc["25%"],
        "50%": desc["50%"],
        "75%": desc["75%"],
        "Max": desc["max"]
    })

summary_df = pd.DataFrame(summary_list)
display(summary_df)
```

### **Bước 7: Feature Engineering cơ bản**

**Mục tiêu:** nhằm tạo ra các biến mới có ý nghĩa từ các cột thời gian để hỗ trợ phân tích. Việc tạo các biến này không chỉ làm giàu thông tin cho dữ liệu mà còn hỗ trợ trực quan hóa và mô hình hóa sau này, đồng thời giúp phát hiện các xu hướng và đặc điểm ẩn trong dữ liệu tai nạn.

```python
# Tách Date thành Month, Day
merged_data['Month'] = merged_data['Date'].dt.month
merged_data['Day'] = merged_data['Date'].dt.day

# Tạo biến Time_period theo Hour
def assign_time_period(hour):
    if pd.isna(hour):
        return None
    if 5 <= hour < 10:
        return 'Sáng sớm'
    elif 10 <= hour < 15:
        return 'Ban ngày'
    elif 15 <= hour < 19:
        return 'Chiều tối'
    elif 19 <= hour < 23:
        return 'Tối'
    else:
        return 'Khuya'

merged_data['Time_period'] = merged_data['Hour'].apply(assign_time_period)

# Tạo biến Season theo Month
def assign_season(month):
    if pd.isna(month):
        return None
    if month in [3, 4, 5]:
        return 'Mùa xuân'
    elif month in [6, 7, 8]:
        return 'Mùa hạ'
    elif month in [9, 10, 11]:
        return 'Mùa thu'
    else:
        return 'Mùa đông'

merged_data['Season'] = merged_data['Month'].apply(assign_season)
```

**Mô tả:**
+ Cột **Date** đã chuẩn hóa, ta tách ra các thành phần gồm năm (**Year**), tháng (**Month**), ngày (**Day**) và ngày trong tuần (**Day_of_week**) để nắm được xu hướng theo từng mốc thời gian.
+ Đồng thời, dựa trên cột **Hour** trích xuất từ cột **Time**, ta phân loại giờ trong ngày thành các khung thời gian có ý nghĩa như “Sáng sớm”, “Ban ngày”, “Chiều tối”, “Tối” và “Khuya”, giúp phân tích tai nạn theo thời điểm trong ngày.
+ Bên cạnh đó, từ cột **Month** còn tạo ra biến **Season** để xác định mùa trong năm, phục vụ các phân tích theo mùa vụ. Có một điểm khác biệt rằng, đó là mùa tại Vương quốc Anh khác với Việt Nam: ở Anh, mùa Xuân bắt đầu từ tháng 3, mùa Hạ từ tháng 6, mùa Thu từ tháng 9 và mùa Đông kết thúc vào tháng 2. Việc xác định mùa theo cách này giúp phân tích các xu hướng tai nạn theo mùa phù hợp với bối cảnh khí hậu và thói quen sinh hoạt tại Anh.

**Để trả lời câu hỏi này, phân tích tiếp theo sẽ tiếp cận vấn đề theo ba góc độ chính, lần lượt xem xét:**

Sau khi xác định rằng phần lớn tai nạn chỉ ở mức độ nhẹ, phân tích theo yếu tố **THỜI GIAN (WHEN)** được thực hiện nhằm kiểm tra sự khác biệt về tần suất va chạm giữa các ngày trong tuần và cuối tuần.

Con số này gấp hơn 3.5 lần so với vùng đứng thứ hai là West Midlands (Tây Midlands, 86.469 vụ).
  + Tuy nhiên, khi nhìn vào các vùng từ Top 2 trở xuống, sự phân bố tai nạn vẫn tuân theo một logic chặt chẽ: các vùng có số vụ cao tiếp theo đều là những trung tâm đô thị lớn (như **Tây Midlands**, **Tây Yorkshire**, **Đại Manchester**). Sự chênh lệch khổng lồ giữa London và phần còn lại cho thấy tầm quan trọng của thủ đô, nhưng sự tập trung tai nạn ở các vùng theo sau cũng khẳng định mật độ dân cư và kinh tế chính là động lực chính thúc đẩy tần suất tai nạn trên mọi khu vực.

Sau khi xác định vùng địa lý có số tai nạn cao nhất là London Metropolitan Police, chúng ta tiếp tục nhìn ở góc độ tính chất khu vực (Urban vs Rural) để xem tai nạn có tập trung ở đô thị hay nông thôn trên toàn quốc. ***Câu hỏi đặt ra: Khi thu nhỏ từ vùng địa lý vào môi trường cụ thể, tai nạn xảy ra nhiều hơn ở khu vực đô thị hay nông thôn?***

```python
from matplotlib.ticker import MaxNLocator, FuncFormatter
# Chuẩn hóa dữ liệu
area_series = area_series.replace({'Unallocated': 'Khác'})

area_counts = area_series.value_counts().astype(int)
area_counts = area_counts.loc[['Urban', 'Rural']]
area_df = area_counts.reset_index()
area_df.columns = ['Area', 'Accident_Count']

total = area_df['Accident_Count'].sum()

plt.figure(figsize=(14,8))

area_order = ['Urban', 'Rural']
custom_palette = {
    'Urban': '#8B0000',
    'Rural': '#FF6B6B'
}

# Vẽ biểu đồ
ax = sns.barplot(
    data=area_df,
    x='Area',
    y='Accident_Count',
    palette=[custom_palette[a] for a in area_order]
)

plt.title("Phân bố tai nạn theo khu vực (Đô thị vs Nông thôn)", fontsize=18)
plt.xlabel("Khu vực", fontsize=15)
plt.ylabel("Số vụ tai nạn", fontsize=15)

plt.gca().yaxis.set_major_locator(MaxNLocator(integer=True))
plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x, p: f"{int(x):,}"))

# Ghi số + tỉ lệ trên đầu cột
for i, row in area_df.iterrows():
    plt.text(
        i,
        row["Accident_Count"] + total * 0.005,
        f"{row['Accident_Count']:,} ({row['Accident_Count']/total:.1%})",
        ha='center',
        fontsize=13
    )

plt.ylim(0, area_df['Accident_Count'].max() * 1.15)
plt.tight_layout()
plt.show()
```

**→ Nhận xét:** Từ biểu đồ chúng ta có thể thấy một sự chênh lệch rõ ràng và không thể phủ nhận giữa môi trường **Đô thị** và **Nông thôn**:
+ Khu vực **Đô thị (Urban)** hoàn toàn chiếm ưu thế về số vụ tai nạn, ghi nhận tới 1.284.345 vụ, chiếm tỷ lệ 62.401% trên tổng số. Con số này cao hơn gấp đôi so với Khu vực **Nông thôn (Rural)**, nơi ghi nhận 773.761 vụ, chiếm 37.594%.
+ Sự chênh lệch lớn này có thể là do tại **Đô thị**, tần suất tai nạn cao là hệ quả trực tiếp của Mật độ và Sự phức tạp của hệ thống đường sá, làm tăng khả năng xảy ra va chạm, dẫn đến tần suất cao tổng thể. Dữ liệu khẳng định Khu vực **Đô thị** là tâm điểm của các vụ tai nạn về mặt tần suất, đòi hỏi nguồn lực phải tập trung vào quản lý tắc nghẽn và giao thông phức hợp tại các thành phố.

Sau khi đã quan sát sự khác biệt về số vụ tai nạn giữa đô thị và nông thôn, ta có thể thấy môi trường xảy ra tai nạn có ảnh hưởng rõ rệt đến tần suất va chạm. Tuy nhiên, “đô thị hay nông thôn” vẫn là một góc nhìn khá rộng. Để hiểu rõ hơn nguyên nhân và bối cảnh, ta cần đi sâu vào đặc điểm hạ tầng giao thông. ***Câu hỏi đặt ra: Trong các loại đường khác nhau, đâu là loại đường ghi nhận số vụ tai nạn cao nhất?***

```python
from matplotlib.ticker import MaxNLocator, FuncFormatter
# Đếm số lượng từng loại đường
road_series = merged_data['Road_Type']
road_counts = road_series.value_counts()
road_df = road_counts.reset_index()
road_df.columns = ['Road_Type', 'Accident_Count']

total = road_df['Accident_Count'].sum()

plt.figure(figsize=(18,8))

road_df = road_df.sort_values('Accident_Count', ascending=False)
colors = sns.color_palette("Set2", n_colors=len(road_df))

ax = sns.barplot(
    data=road_df,
    x='Road_Type',
    y='Accident_Count',
    palette=colors
)

plt.title("Số vụ tai nạn theo loại đường", fontsize=20)
plt.xlabel("Loại đường", fontsize=18)
plt.ylabel("Số vụ tai nạn", fontsize=18)

plt.gca().yaxis.set_major_locator(MaxNLocator(integer=True))
plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x, _: f"{int(x):,}"))

# Ghi số + tỉ lệ
for i, row in road_df.iterrows():
    plt.text(
        i,
        row["Accident_Count"] + total * 0.005,
        f"{row['Accident_Count']:,} ({row['Accident_Count']/total:.2%})",
        ha='center',
        fontsize=14
    )

plt.xticks(rotation=0, fontsize=16)
plt.yticks(fontsize=16)

plt.tight_layout()
plt.show()
```

**→ Nhận xét:** Từ biểu đồ đã xác định rõ ràng nơi các sự cố giao thông xảy ra nhiều nhất, và câu trả lời nằm ở sự đơn giản của hạ tầng:
+ Loại đường **Đường đơn - Single Carriageway** (thường là đường hai chiều chỉ có một làn xe cho mỗi hướng) hoàn toàn áp đảo các loại đường khác về tần suất tai nạn, ghi nhận tới 1.505.053 vụ, chiếm tỷ lệ 73.12% trên tổng số. Các loại đường khác chỉ chiếm tỷ lệ nhỏ: **Dual Carriageway - Đường đôi** (thường có dải phân cách) đứng thứ hai với 17.07% (351.356 vụ), tiếp theo là **Roundabout - Vòng xuyến** với 6.85%.
+ Điều này có nghĩa là hơn 7/10 vụ tai nạn xảy ra trên loại đường này. Cho thấy nguyên nhân chính không hẳn là tốc độ cao, mà là mức độ tiếp xúc trực diện và điểm giao. Đường đơn thường xuyên có các nguy cơ như lấn làn khi vượt, va chạm trực diện, và nhiều giao lộ phức tạp hơn so với **Đường Đôi** hoặc **Đường Cao Tốc** có dải phân cách.

Xác định được loại đường rủi ro nhất giúp khoanh vùng yếu tố hạ tầng, nhưng loại đường chỉ là một phần của bức tranh. ***Điều gì xảy ra khi cùng một loại đường nhưng trong điều kiện thời tiết khác nhau? Đây là lúc ta cần xem xét yếu tố thời tiết để hiểu tai nạn có tăng lên trong mưa, sương mù hay thời tiết xấu hay không?***

```python
from matplotlib.ticker import MaxNLocator, FuncFormatter
weather_series = merged_data['Weather_Conditions'].fillna('Unknown')
weather_counts = weather_series.value_counts()
weather_df = weather_counts.reset_index()
weather_df.columns = ['Weather', 'Accident_Count']
total = weather_df['Accident_Count'].sum()
weather_df = weather_df.sort_values('Accident_Count', ascending=False)

plt.figure(figsize=(16, 8))

sns.barplot(
    x='Accident_Count',
    y='Weather',
    data=weather_df,
    palette='Reds_r',
    order=weather_df['Weather']
)

plt.title("Số vụ tai nạn theo điều kiện thời tiết", fontsize=18, fontweight='bold')
plt.xlabel("Số vụ tai nạn", fontsize=14)
plt.ylabel("Điều kiện thời tiết", fontsize=14)
plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{int(x):,}'))

for i, count in enumerate(weather_df['Accident_Count']):
    percent = (count / total * 100)
    plt.text(
        count + total*0.005,
        i,
        f"{count:,} ({percent:.1f}%)",
        va='center',
        fontsize=11
    )

plt.tight_layout()
plt.show()
```

**→ Nhận xét:** Phân tích số vụ tai nạn theo điều kiện thời tiết cho thấy một nghịch lý đáng chú ý: phần lớn tai nạn xảy ra trong điều kiện thời tiết tốt, chứ không phải khi thời tiết xấu. Cụ thể, điều kiện **Fine no high winds (Thời tiết đẹp, không gió lớn)** hoàn toàn áp đảo, chiếm tới 1.698.133 vụ, tương đương 82.5% tổng số tai nạn. Ngược lại, các điều kiện thời tiết xấu như Mưa, Sương mù, hoặc Tuyết chỉ chiếm tỷ lệ nhỏ: **Raining no high winds (Mưa, không gió lớn)** là điều kiện thời tiết xấu phổ biến nhất, chỉ chiếm 11.6% (238.550 vụ). Bên cạnh đó, các điều kiện cực đoan khác (Tuyết, Sương mù, Mưa + Gió lớn) cộng lại chỉ chiếm dưới 5%. Do đó, có thể khẳng định rằng thời tiết xấu không phải là yếu tố chính gây ra phần lớn tai nạn.


Tương tự như với điều kiện thời tiết, ánh sáng cũng là một yếu tố quan trọng cần xem xét để hiểu nguyên nhân tai nạn. ***Câu hỏi đặt ra là: Điều kiện ánh sáng nào ghi nhận số vụ tai nạn cao nhất?***

```python
from matplotlib.ticker import MaxNLocator, FuncFormatter
light_series = merged_data['Light_Conditions'].fillna('Unknown')
light_counts = light_series.value_counts()
light_df = light_counts.reset_index()
light_df.columns = ['Light', 'Accident_Count']

total_light = light_df['Accident_Count'].sum()
light_df = light_df.sort_values('Accident_Count', ascending=False)

plt.figure(figsize=(16, 8))

# Vẽ horizontal bar chart
sns.barplot(
    x='Accident_Count',
    y='Light',
    data=light_df,
    palette='Oranges_d',
    order=light_df['Light']
)

plt.title("Số vụ tai nạn theo điều kiện ánh sáng", fontsize=20)
plt.xlabel("Số vụ tai nạn", fontsize=16)
plt.ylabel("Điều kiện ánh sáng", fontsize=16)
plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{int(x):,}'))

# Ghi số + phần trăm trên từng cột
for index, row in light_df.iterrows():
    plt.text(
        row['Accident_Count'] + total_light*0.005,
        index,
        f"{row['Accident_Count']:,} ({row['Accident_Count']/total_light:.2%})",
        va='center',
        fontsize=12
    )

plt.tight_layout()
plt.show()
```

**→ Nhận xét:** Có thể xác nhận rõ ràng rằng tai nạn tập trung chủ yếu vào ban ngày chứ không phải ban đêm hay trong điều kiện ánh sáng kém. **Ánh sáng Ban ngày (Daylight)** hoàn toàn áp đảo về tần suất tai nạn ghi nhận tới 1.544.358 vụ chiếm tỷ lệ 75.03% trên tổng số. Điều này có nghĩa là 3/4 vụ tai nạn xảy ra khi tầm nhìn hoàn toàn rõ ràng. Ngược lại tai nạn xảy ra trong **bóng tối (Darkness)** chỉ chiếm một phần tư còn lại. Trong đó **Darkness - lights lit (Trời tối - Đèn đường bật sáng)** là điều kiện tối phổ biến nhất chiếm 18.55% (381.731 vụ) Các điều kiện tối khác (không đèn ánh sáng không rõ ràng) cộng lại chỉ chiếm dưới 6%. Dữ liệu khẳng định nguy cơ tai nạn chủ yếu gắn liền với mật độ và cường độ giao thông ban ngày. Rủi ro không đến từ bóng tối hay tầm nhìn kém, mà phát sinh từ chính sự phức tạp của giao thông và hành vi lái xe chủ quan trong điều kiện ánh sáng hoàn toàn rõ ràng.

Như vậy, thông qua phân tích đơn biến về thời điểm, địa lý, điều kiện thời tiết và ánh sáng, chúng ta đã nhận diện được các yếu tố môi trường chính ảnh hưởng đến số vụ tai nạn. Bước tiếp theo là đi sâu vào mức độ nghiêm trọng của tai nạn, để không chỉ biết tai nạn xảy ra ở đâu và khi nào, mà còn hiểu được tác động và hậu quả của từng vụ tai nạn. ***Câu hỏi đặt ra: Mức độ nghiêm trọng của các vụ tai nạn được ghi nhận như thế nào?***

```python
from matplotlib.ticker import MaxNLocator, FuncFormatter
severity_series = merged_data['Accident_Severity'].fillna('Unknown')
severity_counts = severity_series.value_counts()
severity_df = severity_counts.reset_index()
severity_df.columns = ['Severity', 'Accident_Count']
total_severity = severity_df['Accident_Count'].sum()
severity_df = severity_df.sort_values('Accident_Count', ascending=False)

plt.figure(figsize=(14, 6))

# Vẽ cột dọc
sns.barplot(
    x='Severity',
    y='Accident_Count',
    data=severity_df,
    palette='Reds_d'
)

plt.title("Số vụ tai nạn theo mức độ nghiêm trọng", fontsize=18)
plt.xlabel("Mức độ nghiêm trọng", fontsize=14)
plt.ylabel("Số vụ tai nạn", fontsize=14)

# Ghi số + phần trăm trên cột
for index, row in severity_df.iterrows():
    plt.text(
        index,
        row['Accident_Count'] + total_severity*0.007,
        f"{row['Accident_Count']:,} ({row['Accident_Count']/total_severity:.2%})",
        ha='center',
        fontsize=12
    )

plt.tight_layout()
plt.show()
```

**→ Nhận xét:** một bức tranh rõ ràng và trấn an: phần lớn các vụ tai nạn chỉ gây ra tổn thất nhẹ, trong khi tỷ lệ các vụ nghiêm trọng và tử vong chỉ là thiểu số. Cụ thể, **Tổn thất Nhẹ (Slight)** hoàn toàn áp đảo về tần suất, ghi nhận tới 1.765.472 vụ, chiếm tỷ lệ 85.78% trên tổng số, điều này ngầm giả định rằng đa số các sự cố là va chạm tốc độ thấp xảy ra trong các tình huống giao thông phức tạp (như kẹt xe ở Đô thị hoặc giờ tan tầm). Ngược lại, các vụ tai nạn nghiêm trọng và tử vong chỉ chiếm một tỷ lệ rất nhỏ: **Nghiêm trọng (Serious)** chiếm 12.89% (265.396 vụ), và **Tử vong (Fatal)** ghi nhận tỷ lệ thấp nhất tuyệt đối, chỉ 1.33% (27.328 vụ), tuy chiếm tỷ lệ nhỏ nhưng đại diện cho các vụ va chạm tốc độ cao hơn hoặc liên quan đến sự va đập mạnh.
