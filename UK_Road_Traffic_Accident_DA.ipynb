    + Road_Surface_Conditions: Tình trạng bề mặt đường
    + Special_Conditions_at_Site: Điều kiện đặc biệt tại hiện trường
    + Carriageway_Hazards: Nguy cơ/chướng ngại vật trên đường xe chạy
  + **Nhóm 5: Hậu quả và đáp ứng:**
    + Number_of_Vehicles: Số phương tiện liên quan
    + Number_of_Casualties: Số lượng nạn nhân
    + Did_Police_Officer_Attend_Scene_of_Accident: Cảnh sát có mặt tại hiện trường không?

    + Pedestrian_Crossing-Human_Control: Người điều khiển vạch sang đường
    + Pedestrian_Crossing-Physical_Facilities: Cơ sở vật chất cho người đi bộ
    + InScotland: Tai nạn có xảy ra ở Scotland không

+ **B. Tệp Vehicle_Information (Thông tin phương tiện)**

    + Engine_Capacity_.CC.: Dung tích động cơ
    + Propulsion_Code: Loại động cơ
    + Was_Vehicle_Left_Hand_Drive: Có phải xe tay lái nghịch không

  + **Nhóm 2: Thông tin người lái và hành trình:**

    + Journey_Purpose_of_Driver: Mục đích chuyến đi
    + Driver_Home_Area_Type: Loại khu vực cư trú của tài xế
    + Driver_IMD_Decile: Mức độ thiếu thốn của tài xế
  + **Nhóm 3: Động lực học va chạm:**
    + Vehicle_Manoeuvre: Thao tác của xe trước khi va chạm
    + X1st_Point_of_Impact: Điểm va chạm đầu tiên
    + Vehicle_Leaving_Carriageway: Xe có rời khỏi phần đường xe chạy không
    + Skidding_and_Overturning: Có trượt bánh và lật xe không
    + Hit_Object_in_Carriageway: Vật thể bị va chạm trên làn đường
    + Hit_Object_off_Carriageway: Vật thể bị va chạm ngoài làn đường
    + Junction_Location: Vị trí tại giao lộ
    + Towing_and_Articulation: Tình trạng kéo theo hoặc đầu kéo
    + Vehicle_Location.Restricted_Lane: Vị trí xe trên làn đường bị hạn chế

**A. File Accident_Information**

**Thực hiện xử lí các giá trị thiếu (Missing values) theo quy ước:**

- *Tỉ lệ missing >50%: xóa cột*

- *Tỉ lệ mising 5-50%: impute (mean/median).*

- *Tỉ lệ mising <5%: xóa dòng.*

```python
# Xóa các dòng có missing values
accidents_draft = accidents_draft.dropna(subset=['Time', 'Speed_limit', '1st_Road_Number'])

# Kiểm tra lại missing values
print("Missing values sau khi xóa:")
missing_after = accidents_draft[['Time', 'Speed_limit', '1st_Road_Number']].isnull().sum()
print(missing_after)
```

**- Kiểm tra các giá trị Missing của biến phân loại:**

```python
categorical_cols = accidents_draft.select_dtypes(include=['object']).columns
total_records = len(accidents_draft)

special_found = False

for col in categorical_cols:
    col_series = accidents_draft[col].astype(str).str.strip()
    col_series_lower = col_series.str.lower()

    # Đếm 3 loại special value
    data_missing_count = (col_series == "Data missing or out of range").sum()
    unknown_count = (col_series_lower == "unknown").sum()
    unallocated_count = (col_series_lower == "unallocated").sum()

    # Nếu cột không có special → bỏ qua
    if data_missing_count + unknown_count + unallocated_count == 0:
        continue

    # Print header (chỉ 1 lần)
    if not special_found:
        print(f"{'CỘT':<35} {'GIÁ TRỊ':<40} {'SỐ LƯỢNG':>10}   {'TỶ LỆ':>7}")
        print("-"*100)
        special_found = True

    # In từng giá trị
    if data_missing_count > 0:
        print(f"{col:<35} {'Data missing or out of range':<40} {data_missing_count:>10,}   {data_missing_count/total_records*100:>6.4f}%")

    if unknown_count > 0:
        print(f"{col:<35} {'Unknown':<40} {unknown_count:>10,}   {unknown_count/total_records*100:>6.4f}%")

    if unallocated_count > 0:
        print(f"{col:<35} {'Unallocated':<40} {unallocated_count:>10,}   {unallocated_count/total_records*100:>6.4f}%")

    print("-"*100)
```

**Thực hiện xử lí các giá trị thiếu (Missing values) theo quy ước:**
- *Gom các giá trị không rõ. Bao gồm: Data missing or out of range, Unknown, Other, Other/Not known, Not known  gộp thành một category chung là Missing.*
- *Tỉ lệ missing >5%: Keep (giữ nguyên category Missing)*
- *Tỉ lệ mising <5%: Impute (mode).*

```python
missing_labels = [
    "Data missing or out of range",
    "Unknown"
]
categorical_cols = accidents_draft.select_dtypes(include=['object']).columns
for col in categorical_cols:
    if col == "Urban_or_Rural_Area":
        accidents_draft[col] = accidents_draft[col].replace({
            "Data missing or out of range": "Missing",
            "Unknown": "Missing"
        })
    else:
        accidents_draft[col] = accidents_draft[col].replace(missing_labels, "Missing")

for col in categorical_cols:
    missing_rate = (accidents_draft[col] == "Missing").mean()

    if missing_rate < 0.05:
        # Impute mode
        mode_val = accidents_draft[col].mode()[0]
        accidents_draft[col] = accidents_draft[col].replace("Missing", mode_val)
        print(f"[IMPUTE] {col}: {missing_rate:.2%} → replaced Missing with mode = {mode_val}")
    else:
        print(f"[KEEP] {col}: {missing_rate:.2%} → kept as separate 'Missing' category")
```

```python
# Thống kê chi tiết các cột có giá trị missing
print("- Thống kê chi tiết giá trị missing theo cột:")
missing_stats = vehicle_draft.isnull().sum()
missing_percent = (vehicle_draft.isnull().sum() / len(vehicle_draft)) * 100
missing_df = pd.DataFrame({
    'Số lượng missing': missing_stats,
    'Tỷ lệ %': missing_percent
}).sort_values('Số lượng missing', ascending=False)
missing_df = missing_df[missing_df['Số lượng missing'] > 0]
display(missing_df)
```

- *Tỉ lệ missing >50%: xóa cột*

- *Tỉ lệ mising 5-50%: impute (mean/median).*

- *Tỉ lệ mising <5%: xóa dòng.*

```python
# Điền median cho Age_of_Vehicle
age_median = vehicle_draft['Age_of_Vehicle'].median()
vehicle_draft['Age_of_Vehicle'] = vehicle_draft['Age_of_Vehicle'].fillna(age_median)

# Điền median cho Engine_Capacity_.CC.
engine_median = vehicle_draft['Engine_Capacity_.CC.'].median()
vehicle_draft['Engine_Capacity_.CC.'] = vehicle_draft['Engine_Capacity_.CC.'].fillna(engine_median)

# Điền median cho Driver_IMD_Decile
imd_median = vehicle_draft['Driver_IMD_Decile'].median()
vehicle_draft['Driver_IMD_Decile'] = vehicle_draft['Driver_IMD_Decile'].fillna(imd_median)

print("Missing values sau khi impute:\n")
missing_after = vehicle_draft[['Age_of_Vehicle', 'Engine_Capacity_.CC.', 'Driver_IMD_Decile']].isnull().sum()
print(missing_after)
```

**- Kiểm tra các giá trị Missing của biến phân loại:**

```python
categorical_cols = vehicle_draft.select_dtypes(include=['object']).columns
total_records = len(vehicle_draft)

special_values = [
    "data missing or out of range",
    "unknown",
    "not known",
    "other/not known (2005-10)",
    "other"
]

special_found = False

for col in categorical_cols:

    col_series = vehicle_draft[col].astype(str).str.strip()
    col_series_lower = col_series.str.lower()

    counts = {}

    # Đếm từng loại special value
    for sv in special_values:
        count = (col_series_lower == sv).sum()
        if count > 0:
            counts[sv] = count

    if len(counts) == 0:
        continue

    # In header (chỉ 1 lần)
    if not special_found:
        print(f"\n{'CỘT':<35} {'GIÁ TRỊ':<45} {'SỐ LƯỢNG':>12}   {'TỶ LỆ':>8}")
        print("-" * 105)
        special_found = True

    # In từng loại missing của cột đó
    for sv, count in counts.items():
        percent = (count / total_records) * 100
        print(f"{col:<35} {sv:<45} {count:>12,}   {percent:>7.4f}%")

    print("-" * 105)
```

**Thực hiện xử lí các giá trị thiếu (Missing values) theo quy ước:**
- *Gom các giá trị không rõ. Bao gồm: Data missing or out of range, Unknown, Other, Other/Not known, Not known  gộp thành một category chung là Missing.*
- *Tỉ lệ missing >5%: Keep (giữ nguyên category Missing)*
- *Tỉ lệ mising <5%: Impute (mode).*

```python
missing_labels = [
    "Data missing or out of range",
    "Unknown",
    "Not known",
    "Other",
    "Other/Not known (2005-10)",
    "Unallocated"
]

categorical_cols = vehicle_draft.select_dtypes(include=['object']).columns

for col in categorical_cols:
    vehicle_draft[col] = vehicle_draft[col].astype(str).str.strip()
    vehicle_draft[col] = vehicle_draft[col].replace(
        {label: "Missing" for label in missing_labels}
    )

# Impute
for col in categorical_cols:
    missing_rate = (vehicle_draft[col] == "Missing").mean()

    if missing_rate < 0.05:
        # Impute bằng mode
        mode_val = vehicle_draft[col].mode()[0]
        vehicle_draft[col] = vehicle_draft[col].replace("Missing", mode_val)
        print(f"[IMPUTE] {col}: {missing_rate:.2%} → replaced Missing with mode = {mode_val}")
    else:
        # Giữ Missing
        print(f"[KEEP] {col}: {missing_rate:.2%} → kept as separate 'Missing' category")
```

+ Biến **Vehicle_Reference** (Outlier: 1.85%). Đây là biến tham chiếu phương tiện với phân phối khá tập trung (75% giá trị từ 1-2). Giá trị tối đa 91 có thể đại diện cho các vụ tai nạn đặc biệt nghiêm trọng với nhiều phương tiện liên quan. Với tỷ lệ outlier 1.85% tương đối thấp, giữ nguyên các giá trị này vì có thể chứa thông tin quan trọng về các vụ tai nạn phức tạp.

**Thực hiện xử lí các giá trị Outliers bất thường của 2 biến **Age_of_Vehicle** và **Engine_Capacity_.CC. theo quy ước sử dụng IQR Capping:****
+ lower bound = Q1 - 1.5 * IQR
+ upper bound = Q3 + 1.5 * IQR
+ Giá trị < lower bound → gán bằng lower bound
+ Giá trị > upper bound → gán bằng upper bound

```python
num_cols_cap = ['Age_of_Vehicle', 'Engine_Capacity_.CC.']

summary_list = []

for col in num_cols_cap:
    Q1 = vehicle_draft[col].quantile(0.25)
    Q3 = vehicle_draft[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    print(f"Column: {col}")
    print(f"Lower bound: {lower_bound:.2f}, Upper bound: {upper_bound:.2f}")

    # Capping
    vehicle_draft[col] = np.where(
        vehicle_draft[col] < lower_bound, lower_bound,
        np.where(vehicle_draft[col] > upper_bound, upper_bound, vehicle_draft[col])
    )

    # Lưu summary sau capping để hiển thị trong bảng
    desc = vehicle_draft[col].describe()
    summary_list.append({
        "Variable": col,
        "Count": desc["count"],
        "Mean": desc["mean"],
        "Std": desc["std"],
        "Min": desc["min"],
        "25%": desc["25%"],
        "50%": desc["50%"],
        "75%": desc["75%"],
        "Max": desc["max"]
    })

summary_df = pd.DataFrame(summary_list)
display(summary_df)
```

###**Bước 5: Chuyển đổi kiểu dữ liệu**

**Mục tiêu:** Chuẩn hóa các cột liên quan đến thời gian để phục vụ cho phân

```python
# Chuyển Date từ string → datetime
merged_data['Date'] = pd.to_datetime(merged_data['Date'], errors='coerce')

# Chuyển Time → Hour (numeric)
merged_data['Hour'] = pd.to_datetime(merged_data['Time'], format='%H:%M', errors='coerce').dt.hour
```

**Mô tả:**

+ Đồng thời, cột Time được chuyển thành một cột số Hour trích xuất giờ trong ngày, thuận tiện cho việc phân loại các khung giờ như buổi sáng, buổi trưa, chiều, tối và ban đêm. Việc chuyển đổi này giúp dữ liệu thời gian trở nên nhất quán, giảm rủi ro lỗi và chuẩn bị tốt cho các bước phân tích tiếp theo.

### **Bước 6: Feature Engineering cơ bản**

**Mục tiêu:** nhằm tạo ra các biến mới có ý nghĩa từ các cột thời gian để hỗ trợ phân tích. Việc tạo các biến này không chỉ làm giàu thông tin cho dữ liệu mà còn hỗ trợ trực quan hóa và mô hình hóa sau này, đồng thời giúp phát hiện các xu hướng và đặc điểm ẩn trong dữ liệu tai nạn.

```python
# Tách Date thành Year, Month, Day, Day_of_week
merged_data['Year'] = merged_data['Date'].dt.year
merged_data['Month'] = merged_data['Date'].dt.month
merged_data['Day'] = merged_data['Date'].dt.day
merged_data['Day_of_week'] = merged_data['Date'].dt.dayofweek

# Tạo biến Time_period theo Hour
def assign_time_period(hour):
    if pd.isna(hour):
        return None
    if 5 <= hour < 10:
        return 'Sáng sớm'
    elif 10 <= hour < 15:
        return 'Ban ngày'
    elif 15 <= hour < 19:
        return 'Chiều tối'
    elif 19 <= hour < 23:
        return 'Tối'
    else:
        return 'Khuya'

merged_data['Time_period'] = merged_data['Hour'].apply(assign_time_period)

# Tạo biến Season theo Month
def assign_season(month):
    if pd.isna(month):
        return None
    if month in [3, 4, 5]:
        return 'Mùa xuân'
    elif month in [6, 7, 8]:
        return 'Mùa hạ'
    elif month in [9, 10, 11]:
        return 'Mùa thu'
    else:
        return 'Mùa đông'

merged_data['Season'] = merged_data['Month'].apply(assign_season)
```

**Mô tả:**
+ Cột **Date** đã chuẩn hóa, ta tách ra các thành phần gồm năm (**Year**), tháng (**Month**), ngày (**Day**) và ngày trong tuần (**Day_of_week**) để nắm được xu hướng theo từng mốc thời gian.
+ Đồng thời, dựa trên cột **Hour** trích xuất từ cột **Time**, ta phân loại giờ trong ngày thành các khung thời gian có ý nghĩa như “Sáng sớm”, “Ban ngày”, “Chiều tối”, “Tối” và “Khuya”, giúp phân tích tai nạn theo thời điểm trong ngày.
+ Bên cạnh đó, từ cột **Month** còn tạo ra biến **Season** để xác định mùa trong năm, phục vụ các phân tích theo mùa vụ. Có một điểm khác biệt rằng, đó là mùa tại Vương quốc Anh khác với Việt Nam: ở Anh, mùa Xuân bắt đầu từ tháng 3, mùa Hạ từ tháng 6, mùa Thu từ tháng 9 và mùa Đông kết thúc vào tháng 2. Việc xác định mùa theo cách này giúp phân tích các xu hướng tai nạn theo mùa phù hợp với bối cảnh khí hậu và thói quen sinh hoạt tại Anh.

Sự chênh lệch khổng lồ giữa London và phần còn lại cho thấy tầm quan trọng của thủ đô, nhưng sự tập trung tai nạn ở các vùng theo sau cũng khẳng định mật độ dân cư và kinh tế chính là động lực chính thúc đẩy tần suất tai nạn trên mọi khu vực.

Sau khi xác định vùng địa lý có số tai nạn cao nhất là London Metropolitan Police, chúng ta tiếp tục nhìn ở góc độ tính chất khu vực (Urban vs Rural) để xem tai nạn có tập trung ở đô thị hay nông thôn trên toàn quốc. ***Câu hỏi đặt ra: Khi thu nhỏ từ vùng địa lý vào môi trường cụ thể, tai nạn xảy ra nhiều hơn ở khu vực đô thị hay nông thôn?***

```python
# Chuẩn hóa dữ liệu
area_series = area_series.replace({'Unallocated': 'Khác'})

area_counts = area_series.value_counts().astype(int)
area_counts = area_counts.loc[['Urban', 'Rural']]
area_df = area_counts.reset_index()
area_df.columns = ['Area', 'Accident_Count']

total = area_df['Accident_Count'].sum()

plt.figure(figsize=(14,8))

area_order = ['Urban', 'Rural']
custom_palette = {
    'Urban': '#8B0000',
    'Rural': '#FF6B6B'
}

# Vẽ biểu đồ
ax = sns.barplot(
    data=area_df,
    x='Area',
    y='Accident_Count',
    palette=[custom_palette[a] for a in area_order]
)

plt.title("Phân bố tai nạn theo khu vực (Đô thị vs Nông thôn)", fontsize=18)
plt.xlabel("Khu vực", fontsize=15)
plt.ylabel("Số vụ tai nạn", fontsize=15)

plt.gca().yaxis.set_major_locator(MaxNLocator(integer=True))
plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x, p: f"{int(x):,}"))

# Ghi số + tỉ lệ trên đầu cột
for i, row in area_df.iterrows():
    plt.text(
        i,
        row["Accident_Count"] + total * 0.005,
        f"{row['Accident_Count']:,} ({row['Accident_Count']/total:.1%})",
        ha='center',
        fontsize=13
    )

plt.ylim(0, area_df['Accident_Count'].max() * 1.15)
plt.tight_layout()
plt.show()
```

**→ Nhận xét:** Từ biểu đồ chúng ta có thể thấy một sự chênh lệch rõ ràng và không thể phủ nhận giữa môi trường **Đô thị** và **Nông thôn**:
+ Khu vực **Đô thị (Urban)** hoàn toàn chiếm ưu thế về số vụ tai nạn, ghi nhận tới 1.284.345 vụ, chiếm tỷ lệ 62.401% trên tổng số. Con số này cao hơn gấp đôi so với Khu vực **Nông thôn (Rural)**, nơi ghi nhận 773.761 vụ, chiếm 37.594%.
+ Sự chênh lệch lớn này có thể là do tại **Đô thị**, tần suất tai nạn cao là hệ quả trực tiếp của Mật độ và Sự phức tạp của hệ thống đường sá, làm tăng khả năng xảy ra va chạm, dẫn đến tần suất cao tổng thể. Dữ liệu khẳng định Khu vực **Đô thị** là tâm điểm của các vụ tai nạn về mặt tần suất, đòi hỏi nguồn lực phải tập trung vào quản lý tắc nghẽn và giao thông phức hợp tại các thành phố.

Sau khi đã quan sát sự khác biệt về số vụ tai nạn giữa đô thị và nông thôn, ta có thể thấy môi trường xảy ra tai nạn có ảnh hưởng rõ rệt đến tần suất va chạm. Tuy nhiên, “đô thị hay nông thôn” vẫn là một góc nhìn khá rộng. Để hiểu rõ hơn nguyên nhân và bối cảnh, ta cần đi sâu vào đặc điểm hạ tầng giao thông. ***Câu hỏi đặt ra: Trong các loại đường khác nhau, đâu là loại đường ghi nhận số vụ tai nạn cao nhất?***

```python
# Đếm số lượng từng loại đường
road_counts = road_series.value_counts()
road_df = road_counts.reset_index()
road_df.columns = ['Road_Type', 'Accident_Count']

total = road_df['Accident_Count'].sum()

plt.figure(figsize=(18,8))

road_df = road_df.sort_values('Accident_Count', ascending=False)
colors = sns.color_palette("Set2", n_colors=len(road_df))

ax = sns.barplot(
    data=road_df,
    x='Road_Type',
    y='Accident_Count',
    palette=colors
)

plt.title("Số vụ tai nạn theo loại đường", fontsize=20)
plt.xlabel("Loại đường", fontsize=18)
plt.ylabel("Số vụ tai nạn", fontsize=18)

plt.gca().yaxis.set_major_locator(MaxNLocator(integer=True))
plt.gca().yaxis.set_major_formatter(FuncFormatter(lambda x, _: f"{int(x):,}"))

# Ghi số + tỉ lệ
for i, row in road_df.iterrows():
    plt.text(
        i,
        row["Accident_Count"] + total * 0.005,
        f"{row['Accident_Count']:,} ({row['Accident_Count']/total:.2%})",
        ha='center',
        fontsize=14
    )

plt.xticks(rotation=0, fontsize=16)
plt.yticks(fontsize=16)

plt.tight_layout()
plt.show()
```

**→ Nhận xét:** Từ biểu đồ đã xác định rõ ràng nơi các sự cố giao thông xảy ra nhiều nhất, và câu trả lời nằm ở sự đơn giản của hạ tầng:
+ Loại đường **Đường đơn - Single Carriageway** (thường là đường hai chiều chỉ có một làn xe cho mỗi hướng) hoàn toàn áp đảo các loại đường khác về tần suất tai nạn, ghi nhận tới 1.505.053 vụ, chiếm tỷ lệ 73.12% trên tổng số. Các loại đường khác chỉ chiếm tỷ lệ nhỏ: **Dual Carriageway - Đường đôi** (thường có dải phân cách) đứng thứ hai với 17.07% (351.356 vụ), tiếp theo là **Roundabout - Vòng xuyến** với 6.85%.
+ Điều này có nghĩa là hơn 7/10 vụ tai nạn xảy ra trên loại đường này. Cho thấy nguyên nhân chính không hẳn là tốc độ cao, mà là mức độ tiếp xúc trực diện và điểm giao. Đường đơn thường xuyên có các nguy cơ như lấn làn khi vượt, va chạm trực diện, và nhiều giao lộ phức tạp hơn so với **Đường Đôi** hoặc **Đường Cao Tốc** có dải phân cách.
