### **1.2. Dữ liệu vào (Input Data)**
+ **A. Tệp Accident_Information (Thông tin tai nạn):**
  + **Nhóm 1: Thông tin thời gian:**
    + Date: Ngày xảy ra tai nạn.
    + Time: Thời gian xảy ra tai nạn.
    + Day_of_Week: Thứ trong tuần.
    + Year: Năm xảy ra tai nạn.
  + **Nhóm 2: Thông tin địa lí và vị trí:**
    + Latitude: Vĩ độ.
    + Longitude: Kinh độ.
    + Location_Easting_OSGR: Tọa độ đông OSGR.
    + Location_Northing_OSGR: Tọa độ bắc OSGR.
    + Urban_or_Rural_Area: Khu vực thành thị hay nông thôn.
    + Local_Authority_(District): Khu vực hành chính địa phương.
    + Local_Authority_(Highway): Cơ quan quản lý đường bộ địa phương.
    + LSOA_of_Accident_Location: Khu vực thống kê của vị trí tai nạn.
  + **Nhóm 3: Đặc điểm đường và giao lộ:**
    + 1st_Road_Class, 1st_Road_Number: Hạng và số hiệu đường chính
    + 2nd_Road_Class, 2nd_Road_Number: Hạng và số hiệu đường thứ hai
    + Road_Type: Loại đường
    + Speed_Limit: Tốc độ giới hạn
    + Junction_Control, Junction_Detail: Cách kiểm soát và chi tiết giao lộ
  + **Nhóm 4: Điều kiện môi trường và hiện trường:**
    + Light_Conditions: Điều kiện ánh sáng
    + Weather_Conditions: Điều kiện thời tiết
    + Road_Surface_Conditions: Tình trạng bề mặt đường
    + Special_Conditions_at_Site: Điều kiện đặc biệt tại hiện trường
    + Carriageway_Hazards: Nguy cơ/chướng ngại vật trên đường xe chạy
  + **Nhóm 5: Hậu quả và đáp ứng:**
    + Number_of_Vehicles: Số phương tiện liên quan
    + Number_of_Casualties: Số lượng nạn nhân
    + Did_Police_Officer_Attend_Scene_of_Accident: Cảnh sát có mặt tại hiện trường không?
    + Police_Force: Vùng Địa lý mà Lực lượng Cảnh sát đó phụ trách quản lý
    + Pedestrian_Crossing-Human_Control: Người điều khiển vạch sang đường
    + Pedestrian_Crossing-Physical_Facilities: Cơ sở vật chất cho người đi bộ
    + InScotland: Tai nạn có xảy ra ở Scotland không

+ **B. Tệp Vehicle_Information (Thông tin phương tiện)**

  + **Nhóm 1: Thông tin phương tiện:**
    + Vehicle_Type: Loại xe
    + make, model: Hãng xe và mẫu xe
    + Age_of_Vehicle: Tuổi của xe
    + Engine_Capacity_.CC.: Dung tích động cơ
    + Propulsion_Code: Loại động cơ
    + Was_Vehicle_Left_Hand_Drive: Có phải xe tay lái nghịch không
    + Year: Năm sản xuất phương tiện
  + **Nhóm 2: Thông tin người lái và hành trình:**
    + Sex_of_Driver: Giới tính của tài xế
    + Age_Band_of_Driver: Nhóm tuổi của tài xế
    + Journey_Purpose_of_Driver: Mục đích chuyến đi
    + Driver_Home_Area_Type: Loại khu vực cư trú của tài xế
    + Driver_IMD_Decile: Mức độ thiếu thốn của tài xế
  + **Nhóm 3: Động lực học va chạm:**
    + Vehicle_Manoeuvre: Thao tác của xe trước khi va chạm
    + X1st_Point_of_Impact: Điểm va chạm đầu tiên
    + Vehicle_Leaving_Carriageway: Xe có rời khỏi phần đường xe chạy không
    + Skidding_and_Overturning: Có trượt bánh và lật xe không
    + Hit_Object_in_Carriageway: Vật thể bị va chạm trên làn đường
    + Hit_Object_off_Carriageway: Vật thể bị va chạm ngoài làn đường
    + Junction_Location: Vị trí tại giao lộ
    + Towing_and_Articulation: Tình trạng kéo theo hoặc đầu kéo
    + Vehicle_Location.Restricted_Lane: Vị trí xe trên làn đường bị hạn chế

###**Bước 3: Hợp nhất dữ liệu**

**- Do mỗi vụ tai nạn có thể liên quan đến nhiều phương tiện khác nhau, bảng Vehicle cần được tổng hợp theo khóa Accident_Index để đảm bảo mỗi bản ghi trong dataset cuối cùng đại diện cho một vụ tai nạn duy nhất, tránh hiện tượng nhân bản bản ghi khi hợp nhất dữ liệu, đồng thời vẫn giữ được thông tin đặc trưng của các phương tiện liên quan đến mỗi vụ tai nạn. Quá trình tổng hợp, sẽ áp dụng theo quy ước:**

  - Đối với các biến phân loại (categorical):
Giá trị đại diện được xác định bằng mode (giá trị xuất hiện nhiều nhất), nhằm phản ánh đặc điểm phổ biến nhất của các phương tiện liên quan trong cùng một vụ tai nạn.
  - Đối với các biến dạng số (numerical):
Sử dụng median (thay vì mean) vì nó không bị ảnh hưởng bởi các giá trị ngoại lai, đồng thời phản ánh xu hướng trung tâm ổn định hơn của dữ liệu.

```python
import polars as pl

# Đổi tên cột để sử dụng
vehicle_clean1 = vehicle_clean1.rename(columns={'Engine_Capacity_.CC.':'Engine_Capacity_CC'})
df = pl.from_pandas(vehicle_clean1)

# Aggregate theo Accident_Index
vehicle_agg_pl = df.group_by("Accident_Index").agg([
    pl.col("Vehicle_Type").mode().first().alias("Vehicle_Type"),
    pl.col("Vehicle_Manoeuvre").mode().first().alias("Vehicle_Manoeuvre"),
    pl.col("Vehicle_Leaving_Carriageway").mode().first().alias("Vehicle_Leaving_Carriageway"),
    pl.col("Age_Band_of_Driver").mode().first().alias("Age_Band_of_Driver"),
    pl.col("Sex_of_Driver").mode().first().alias("Sex_of_Driver"),
    pl.col("Age_of_Vehicle").median().alias("Age_of_Vehicle"),
    pl.col("Engine_Capacity_CC").median().alias("Engine_Capacity_CC")
])

vehicle_agg = vehicle_agg_pl.to_pandas()
print(f"Hoàn thành! Kết quả có {len(vehicle_agg)} dòng")
```

**- Sau khi đã tổng hợp dữ liệu phương tiện theo Accident_Index, chúng ta có bảng vehicle_agg với mỗi dòng đại diện cho một vụ tai nạn duy nhất đồng thời chứa thông tin tổng hợp của các phương tiện liên quan.**

**- Bước tiếp theo, chúng ta sẽ gộp bảng Accident với bảng Vehicle đã tổng hợp để tạo dataset cuối cùng phục vụ cho phân tích khám phá dữ liệu (EDA) và xây dựng mô hình. Với Khóa chính để merge là Accident_Index, sử dụng phương pháp LEFT JOIN để đảm bảo giữ lại tất cả các vụ tai nạn trong accidents_clean kể cả những vụ không có thông tin phương tiện tương ứng.**

```python
merged_data = pd.merge(
    accidents_clean1, # bảng tai nạn
    vehicle_agg,  # bảng vehicle đã aggregate
    on='Accident_Index',
    how='left'
)

merged_data.info()
display(merged_data.head(5))
```

###**Bước 4: Xử lí dữ liệu thiếu (Missing values)**

**- Kiểm tra các giá trị thiếu (Missing Values) của biến số:**

```python
# Lọc các biến số
numerical_cols = merged_data.select_dtypes(include=['float64','int64']).columns

# Thống kê missing chỉ cho biến số
missing_stats = merged_data[numerical_cols].isnull().sum()
missing_percent = (missing_stats / len(merged_data)) * 100

missing_df = pd.DataFrame({
    'Số lượng missing': missing_stats,
    'Tỷ lệ %': missing_percent
}).sort_values('Số lượng missing', ascending=False)

# Chỉ hiển thị những cột có missing
missing_df = missing_df[missing_df['Số lượng missing'] > 0]

display(missing_df)
```

**- Kiểm tra các giá trị thiếu (Missing Values) của biến phân loại:**

```python
# Xác định các biến phân loại
categorical_cols = merged_data.select_dtypes(include=['object']).columns

# Kiểm tra missing values cho biến phân loại
categorical_missing = merged_data[categorical_cols].isnull().sum()
categorical_missing_percent = (categorical_missing / len(merged_data)) * 100
categorical_missing_df = pd.DataFrame({
    'Số lượng missing': categorical_missing,
    'Tỷ lệ %': categorical_missing_percent
}).sort_values('Số lượng missing', ascending=False)
categorical_missing_df = categorical_missing_df[categorical_missing_df['Số lượng missing'] > 0]
display(categorical_missing_df)
```

**Thực hiện xử lí các giá trị thiếu (Missing values) cho cả biến số và biến phân loại theo quy ước:**

- Tỉ lệ missing >50%: xóa cột

- Tỉ lệ mising 5-50%: impute (median) nếu là biến số, impute nhãn (Missing) nếu là biến phân loại.

- Tỉ lệ mising <5%: xóa dòng.

```python
# Xử lý biến số
for col in numerical_cols:
    missing_rate = merged_data[col].isnull().mean()

    if missing_rate == 0:
        continue  # Bỏ qua cột không missing

    if missing_rate > 0.5:
        merged_data.drop(columns=[col], inplace=True)
        print(f"[DROP COL] {col}: missing rate = {missing_rate:.5%} >50%, drop column")

    elif 0.05 <= missing_rate <= 0.5:
        median_val = merged_data[col].median()
        merged_data[col].fillna(median_val, inplace=True)
        print(f"[IMPUTE] {col}: missing rate = {missing_rate:.5%}, fillna with median = {median_val}")

    else:  # missing_rate <5%
        merged_data = merged_data[merged_data[col].notnull()]
        print(f"[DROP ROW] {col}: missing rate = {missing_rate:.5%} <5%, drop rows with NaN")

# Xử lý biến phân loại
for col in categorical_cols:
    missing_rate = merged_data[col].isnull().mean()

    if missing_rate == 0:
        continue  # Bỏ qua cột không missing

    if missing_rate > 0.5:
        merged_data.drop(columns=[col], inplace=True)
        print(f"[DROP COL] {col}: missing rate = {missing_rate:.5%} >50%, drop column")

    elif 0.05 <= missing_rate <= 0.5:
        merged_data[col].fillna('Missing', inplace=True)
        print(f"[IMPUTE] {col}: missing rate = {missing_rate:.5%}, fillna with 'Missing'")

    else:  # missing_rate <5%
        merged_data = merged_data[merged_data[col].notnull()]
        print(f"[DROP ROW] {col}: missing rate = {missing_rate:.5%} <5%, drop rows with NaN")
```

**- Tiếp theo, thực hiện kiểm tra các giá trị không rõ ràng của biến phân loại:**

```python
categorical_cols = merged_data.select_dtypes(include=['object']).columns
total_records = len(merged_data)

special_values = ["Data missing or out of range", "Unknown", "Unallocated", "Other", "Other/Not known", "Not known"]

special_found = False

for col in categorical_cols:
    col_series = merged_data[col].astype(str).str.strip()
    col_series_lower = col_series.str.lower()

    # Đếm từng giá trị đặc biệt
    counts = {}
    for val in special_values:
        if val.lower() in ["unknown", "unallocated"]:
            counts[val] = (col_series_lower == val.lower()).sum()
        else:
            counts[val] = (col_series == val).sum()

    if sum(counts.values()) == 0:
        continue

    if not special_found:
        print(f"{'CỘT':<35} {'GIÁ TRỊ':<40} {'SỐ LƯỢNG':>10}   {'TỶ LỆ':>7}")
        print("-"*100)
        special_found = True

    # In từng giá trị
    for val, count in counts.items():
        if count > 0:
            print(f"{col:<35} {val:<40} {count:>10,}   {count/total_records*100:>6.4f}%")

    print("-"*100)
```

**- Thực hiện gom các giá trị này thành một category chung (Missing). Bao gồm các giá trị: Data missing or out of range, Unknown, Other, Other/Not known, Not known. Sau khi gom, xử lý theo tỉ lệ missing:**

- Nếu tỉ lệ "Missing" >5%: [KEEP] giữ nguyên category "Missing".
- Nếu tỉ lệ "Missing" <5%: [IMPUTE] mode - giá trị xuất hiện nhiều nhất.

```python
missing_labels = [
    "Data missing or out of range",
    "Unknown",
    "Not known",
    "Other",
    "Other/Not known (2005-10)",
    "Unallocated"
]

categorical_cols = merged_data.select_dtypes(include=['object']).columns

for col in categorical_cols:
    # Chuẩn hóa text
    merged_data[col] = merged_data[col].astype(str).str.strip()

    # Gom các nhãn đặc biệt thành 'Missing'
    merged_data[col] = merged_data[col].replace(
        {label: "Missing" for label in missing_labels}
    )

    # Tính tỉ lệ Missing
    missing_rate = (merged_data[col] == "Missing").mean()

    if missing_rate > 0.05:
        # Giữ nguyên category Missing
        print(f"[KEEP] {col}: {missing_rate:.2%} → kept as separate 'Missing' category")
    elif missing_rate > 0:
        # Impute bằng mode
        mode_val = merged_data[col].mode()[0]
        merged_data[col] = merged_data[col].replace("Missing", mode_val)
        print(f"[IMPUTE] {col}: {missing_rate:.2%} → replaced Missing with mode = {mode_val}")
    # Nếu missing_rate == 0, không làm gì
```

**- Dùng biểu đồ Box Plot để trực quan các biến có Outliers:**

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

num_cols = ['Number_of_Casualties', 'Number_of_Vehicles', 'Age_of_Vehicle', 'Engine_Capacity_CC']

# Set style đẹp
sns.set_style("whitegrid")
sns.set_palette("husl")

# Tạo subplot 2x2
fig, axes = plt.subplots(2, 2, figsize=(16, 8))
axes = axes.flatten()

for idx, (ax, col) in enumerate(zip(axes, num_cols)):
    # Tính toán outlier
    Q1 = merged_data[col].quantile(0.25)
    Q3 = merged_data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = merged_data[
        (merged_data[col] < lower_bound) | (merged_data[col] > upper_bound)
    ].shape[0]

    total_count = merged_data.shape[0]
    outlier_percentage = (outlier_count / total_count) * 100

    # Vẽ boxplot ngang
    sns.boxplot(x=merged_data[col],
                ax=ax,
                orient='h',
                width=0.6,
                fliersize=5,
                linewidth=1.5,
                color=f'C{idx}')

    ax.set_title(col, fontsize=14, fontweight='bold', pad=12, color='#2C3E50')
    ax.set_xlabel('Value', fontsize=11)
    ax.set_ylabel('')

    # Thông tin outlier - đặt ở góc trên phải
    ax.text(0.98, 0.95, f'Outlier: {outlier_percentage:.2f}%',
            transform=ax.transAxes, fontsize=11, fontweight='bold',
            verticalalignment='top', horizontalalignment='right',
            bbox=dict(boxstyle='round,pad=0.4', facecolor='white',
                     edgecolor=f'C{idx}', alpha=0.9, linewidth=1.5))

    stats_text = f'Q1: {Q1:.1f} | Q3: {Q3:.1f}\nIQR: {IQR:.1f}'
    ax.text(0.02, 0.95, stats_text,
            transform=ax.transAxes, fontsize=9,
            verticalalignment='top', horizontalalignment='left',
            bbox=dict(boxstyle='round,pad=0.3', facecolor='whitesmoke', alpha=0.7))

    ax.grid(True, linestyle='--', alpha=0.3, axis='x')
    ax.set_axisbelow(True)

    if col == 'Engine_Capacity_CC':
        ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{int(x):,}'))

plt.tight_layout(pad=3.0)
fig.suptitle('Horizontal Boxplot Analysis with Outlier Detection',
             fontsize=16, fontweight='bold', y=1.02)
plt.show()
```

**Thực hiện xử lí các giá trị Outliers bất thường của 2 biến **Age_of_Vehicle** và **Engine_Capacity_.CC. theo quy ước sử dụng IQR Capping:****
+ lower bound = Q1 - 1.5 * IQR
+ upper bound = Q3 + 1.5 * IQR
+ Giá trị < lower bound → gán bằng lower bound
+ Giá trị > upper bound → gán bằng upper bound

```python
num_cols_cap = ['Age_of_Vehicle', 'Engine_Capacity_CC']

summary_list = []

for col in num_cols_cap:
    Q1 = merged_data[col].quantile(0.25)
    Q3 = merged_data[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    print(f"Column: {col}")
    print(f"Lower bound: {lower_bound:.2f}, Upper bound: {upper_bound:.2f}")

    # Capping
    merged_data[col] = np.where(
        merged_data[col] < lower_bound, lower_bound,
        np.where(merged_data[col] > upper_bound, upper_bound, merged_data[col])
    )

    # Lưu summary sau capping để hiển thị trong bảng
    desc = merged_data[col].describe()
    summary_list.append({
        "Variable": col,
        "Count": desc["count"],
        "Mean": desc["mean"],
        "Std": desc["std"],
        "Min": desc["min"],
        "25%": desc["25%"],
        "50%": desc["50%"],
        "75%": desc["75%"],
        "Max": desc["max"]
    })

summary_df = pd.DataFrame(summary_list)
display(summary_df)
```

### **Bước 7: Feature Engineering cơ bản**

**Mục tiêu:** nhằm tạo ra các biến mới có ý nghĩa từ các cột thời gian để hỗ trợ phân tích. Việc tạo các biến này không chỉ làm giàu thông tin cho dữ liệu mà còn hỗ trợ trực quan hóa và mô hình hóa sau này, đồng thời giúp phát hiện các xu hướng và đặc điểm ẩn trong dữ liệu tai nạn.

```python
# Tách Date thành Month, Day
merged_data['Month'] = merged_data['Date'].dt.month
merged_data['Day'] = merged_data['Date'].dt.day

# Tạo biến Time_period theo Hour
def assign_time_period(hour):
    if pd.isna(hour):
        return None
    if 5 <= hour < 10:
        return 'Sáng sớm'
    elif 10 <= hour < 15:
        return 'Ban ngày'
    elif 15 <= hour < 19:
        return 'Chiều tối'
    elif 19 <= hour < 23:
        return 'Tối'
    else:
        return 'Khuya'

merged_data['Time_period'] = merged_data['Hour'].apply(assign_time_period)

# Tạo biến Season theo Month
def assign_season(month):
    if pd.isna(month):
        return None
    if month in [3, 4, 5]:
        return 'Mùa xuân'
    elif month in [6, 7, 8]:
        return 'Mùa hạ'
    elif month in [9, 10, 11]:
        return 'Mùa thu'
    else:
        return 'Mùa đông'

merged_data['Season'] = merged_data['Month'].apply(assign_season)
```

**Mô tả:**
+ Cột **Date** đã chuẩn hóa, ta tách ra các thành phần gồm năm (**Year**), tháng (**Month**), ngày (**Day**) và ngày trong tuần (**Day_of_week**) để nắm được xu hướng theo từng mốc thời gian.
+ Đồng thời, dựa trên cột **Hour** trích xuất từ cột **Time**, ta phân loại giờ trong ngày thành các khung thời gian có ý nghĩa như “Sáng sớm”, “Ban ngày”, “Chiều tối”, “Tối” và “Khuya”, giúp phân tích tai nạn theo thời điểm trong ngày.
+ Bên cạnh đó, từ cột **Month** còn tạo ra biến **Season** để xác định mùa trong năm, phục vụ các phân tích theo mùa vụ. Có một điểm khác biệt rằng, đó là mùa tại Vương quốc Anh khác với Việt Nam: ở Anh, mùa Xuân bắt đầu từ tháng 3, mùa Hạ từ tháng 6, mùa Thu từ tháng 9 và mùa Đông kết thúc vào tháng 2. ### **6.2. Logistic Regression**

#### Lý thuyết

1. XGBoost là gì?

    XGBoost là viết tắt của eXtreme Gradient Boosting, một thuật toán học máy mạnh mẽ dựa trên mô hình cây quyết định (Decision Tree) và kỹ thuật Boosting.


2. Cơ chế hoạt động

    Quy trình học của XGBoost diễn ra tuần tự:

    * Cây 1: Đưa ra dự đoán ban đầu.

    * Cây 2: Học cách dự đoán phần sai số (residual) mà Cây 1 chưa giải quyết được.

    * Cây 3: Tiếp tục học phần sai số còn lại của tổng hợp (Cây 1 + Cây 2).

    Kết quả: Dự đoán cuối cùng là tổng đóng góp của tất cả các cây.

3. Biểu diễn toán học của XGBoost

  * **Dạng tổng của các cây**

    Mô hình XGBoost được biểu diễn dưới dạng tổng của K cây quyết định:

  $$
  \hat{y}_i = \sum_{k=1}^{K} f_k(x_i)
  $$

  Chú thích:


  $x_i$: dữ liệu đầu vào

  $\hat{y}_i$: giá trị dự đoán cuối cùng

  $f_k$: cây thứ $k$

  $f_k(x_i)$: giá trị dự đoán của cây thứ k

  $k$: số lượng cây trong mô hình

  * **Hàm mục tiêu (Objective Function)**

    Tại vòng lặp huấn luyện thứ t, XGBoost tìm cây mới $f_t$ sao cho hàm mục tiêu nhỏ nhất:

  $$
  Obj^{(t)} = \sum_{i=1}^{n}
  \underbrace{L(y, \hat{y})}_{\text{Loss Function}}
  +
  \underbrace{\Omega(f_t)}_{\text{Regularization}}
  $$

  Chú thích:

  n: số lượng mẫu dữ liệu

  $L(y, \hat{y})$: hàm mất mát, đo độ sai lệch giữa giá trị thực tế($y$) và dự đoán($\hat{y}$).

  $\Omega(f_t)$: hàm phạt độ phức tạp của cây

  * **Hàm phạt (Regularization Term)**

    Hàm phạt trong XGBoost được định nghĩa như sau:

  $$
  \Omega(f) = \gamma T + \frac{1}{2} \lambda \lVert w \rVert^2
  $$

  Chú thích:

  $T$: số lượng lá của cây

  $\gamma$ (Gamma - Hệ số phạt lá):Là "ngưỡng tối thiểu" để được phép chia nhánh. (Cơ chế: Khi mô hình định chia một nhánh thành hai lá mới, nếu việc chia này không làm giảm Loss (sai số) được một lượng lớn hơn $\gamma$, thì XGBoost sẽ cắt bỏ nhánh đó (Pruning))

  $w$: vector trọng số tại các lá, là giá trị điểm số (score) mà lá đó dự đoán.(Ví dụ: Một lá dự đoán "Xác suất tử vong rất cao" có thể có $w = +5$. Lá dự đoán thấp có $w = -5$)

  $\lambda$: Là mức độ nghiêm khắc của hình phạt.(Cơ chế: $\lambda$ càng lớn $\rightarrow$ ép các giá trị $w$ phải nhỏ lại gần bằng 0.Tác dụng: Làm "mượt" dự báo. Ngăn chặn việc một lá nào đó có quyền lực quá lớn (ví dụ: chỉ vì 1 mẫu nhiễu mà $w$ vọt lên +100), giúp mô hình ổn định hơn trước dữ liệu lạ.)

  $\lVert w \rVert^2$: chuẩn L2 của trọng số lá, là tổng bình phương của tất cả các trọng số $w$ trong cây ($\sum w_j^2$).

  Càng nhiều lá, trọng số càng lớn -> Phạt càng nặng


  Ý nghĩa:

  Cây càng nhiều lá → mô hình càng phức tạp → mức phạt càng lớn.

  Giúp hạn chế hiện tượng overfitting (học vẹt), khi mô hình ghi nhớ dữ liệu huấn luyện thay vì học quy luật chung.

Kết luận

* XGBoost là thuật toán học máy dựa trên việc cộng dồn nhiều cây quyết định, trong đó mỗi cây mới được xây dựng để sửa lỗi của các cây trước.

    Dữ liệu tai nạn giao thông có phân bố lớp không cân bằng rõ rệt, trong đó:

    * Tai nạn nhẹ (Slight) chiếm đa số.

    * Tai nạn nghiêm trọng (Serious) và đặc biệt là tử vong (Fatal) chiếm tỷ lệ rất nhỏ.

    Để hạn chế việc mô hình thiên lệch về lớp đa số, sample weight được tính bằng: compute_sample_weight(class_weight='balanced', y=y_train)


    Cách tiếp cận này giúp:

    * Tăng mức phạt khi mô hình dự đoán sai các lớp hiếm.

    * Cải thiện khả năng nhận diện các trường hợp Serious và Fatal, vốn có ý nghĩa quan trọng trong phân tích rủi ro giao thông.


3. Lựa chọn mô hình XGBoost

    Mô hình XGBClassifier được sử dụng do:

    * Khả năng xử lý tốt dữ liệu lớn và phi tuyến.

    * Hiệu quả cao với dữ liệu hỗn hợp gồm biến số và biến phân loại.

    * Phù hợp với bài toán phân loại đa lớp (3 mức độ tai nạn).

    * Cho phép mô hình dự đoán xác suất cho từng lớp, phục vụ cho các phân tích sâu hơn như PDP.

    Xử lý trực tiếp biến phân loại (enable_categorical=True): Giúp XGBoost hiểu và xử lý các cột dữ liệu dạng danh mục (category) ngay lập tức. Việc này giúp loại bỏ bước One-Hot Encoding thủ công, từ đó giảm số chiều dữ liệu và tăng tốc độ tính toán.

    Tối ưu hóa tốc độ huấn luyện (tree_method='hist'): Sử dụng thuật toán dựa trên biểu đồ tần suất (Histogram-based) để tìm điểm chia nhanh hơn. Đây là cấu hình tối ưu cho các tập dữ liệu lớn, giúp tiết kiệm bộ nhớ và giảm đáng kể thời gian chạy mô hình.

    Đảm bảo tính tái lập kết quả (random_state=42): Đảm bảo kết quả mô hình luôn ổn định và giống hệt nhau trong mọi lần chạy thử nghiệm.
4. Phương pháp và Thước đo Đánh giá Mô hình (Model Evaluation Metrics)

    Để đánh giá hiệu quả của mô hình phân loại mức độ tai nạn giao thông (Slight, Serious, Fatal) trong bối cảnh dữ liệu mất cân bằng nghiêm trọng, nghiên cứu không chỉ dựa vào độ chính xác tổng thể (Accuracy) mà sử dụng bộ chỉ số đa chiều bao gồm: Precision, Recall, F1-score và các phương pháp tính trung bình (Macro/Weighted Average).

    Dưới đây là chi tiết các thước đo được sử dụng:

    * Precision (Độ chính xác của dự báo): Precision đo lường mức độ tin cậy của các cảnh báo mà mô hình đưa ra. Nó trả lời câu hỏi: "Trong số các vụ tai nạn mô hình dự báo là Fatal, có bao nhiêu vụ là thật?"$$\text{Precision} = \frac{TP}{TP + FP}$$Ý nghĩa trong bài toán: Precision cao giúp giảm thiểu báo động giả. Tuy nhiên, nếu chỉ tối ưu Precision, mô hình có thể trở nên quá "thận trọng" và bỏ sót các tai nạn thực tế.


    * Recall (Độ nhạy / Tỷ lệ phát hiện): Recall đo lường khả năng "bao phủ" của mô hình. Nó trả lời câu hỏi: "Trong thực tế có 100 vụ tai nạn Fatal, mô hình tìm ra được bao nhiêu vụ?"$$\text{Recall} = \frac{TP}{TP + FN}$$Ý nghĩa trong bài toán: Đây là chỉ số quan trọng hàng đầu trong an toàn giao thông.Mục tiêu là không được bỏ sót các vụ tai nạn nghiêm trọng.Nguyên tắc: Thà báo nhầm (False Positive - Chấp nhận Precision thấp hơn) còn hơn bỏ sót (False Negative - Recall thấp).

    * F1-Score (Điểm cân bằng)F1-score là trung bình điều hòa giữa Precision và Recall, dùng để tìm điểm cân bằng giữa độ tin cậy và độ nhạy.$$\text{F1-score} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$$Ý nghĩa: F1-score đặc biệt hữu ích khi cần đánh giá tổng quát hiệu năng của mô hình trên dữ liệu mất cân bằng, tránh trường hợp một chỉ số quá cao còn chỉ số kia quá thấp.

    * Accuracy (Độ chính xác tổng thể)Accuracy là tỷ lệ giữa tổng số dự đoán đúng trên tổng số mẫu dữ liệu.$$\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}$$
    Trong dữ liệu tai nạn (nơi tai nạn nhẹ chiếm >85%), Accuracy dễ gây ra "Nghịch lý độ chính xác". Một mô hình tồi (chỉ đoán toàn bộ là Nhẹ) vẫn có thể đạt Accuracy rất cao nhưng vô dụng. Do đó, chỉ số này chỉ dùng để tham khảo.

    * Các phương pháp tính trung bình (Averaging Methods)
    Vì đây là bài toán phân loại đa lớp (Multi-class), ta cần tổng hợp các chỉ số trên thành một con số đại diện:
      * Macro Average (Trung bình vĩ mô):Công thức: Tính chỉ số cho từng lớp rồi chia đều:
    $$\frac{(Score_{Slight} + Score_{Serious} + Score_{Fatal})}{3}$$
    Ý nghĩa: Coi trọng lớp thiểu số (Fatal) ngang bằng với lớp đa số (Slight). Nếu Macro Avg thấp, nghĩa là mô hình đang thất bại trong việc dự báo các lớp hiếm. Đây là thước đo khắt khe và công bằng nhất.
      * Weighted Average (Trung bình trọng số): Tính chỉ số cho từng lớp, mỗi lớp được nhân với số lượng mẫu thực tế của lớp đó, lớp nào nhiều dữ liệu hơn → ảnh hưởng nhiều hơn đến kết quả cuối. Ý nghĩa: Phản ánh hiệu suất trên dòng dữ liệu thực tế. Chỉ số này thường cao hơn Macro Avg do được "kéo điểm" bởi lớp đa số (Slight).

    * Confusion Matrix (Ma trận nhầm lẫn)Confusion Matrix là bảng đối chiếu chi tiết giữa Dự đoán và Thực tế, giúp phân tích hành vi cụ thể của mô hình.$$\begin{array}{c|ccc}
    \text{Thực tế \ Dự đoán} & \text{Slight} & \text{Serious} & \text{Fatal} \\
    \hline
    \text{Slight} & \mathbf{TP_{Slight}} & E_{12} & E_{13} \\
    \text{Serious} & E_{21} & \mathbf{TP_{Serious}} & E_{23} \\
    \text{Fatal} & \mathbf{FN_{Fatal}} & E_{32} & \mathbf{TP_{Fatal}} \\
    \end{array}$$Mục tiêu phân tích:Quan sát đường chéo chính ($\mathbf{TP}$): Số lượng dự đoán đúng.Quan sát $FN_{Fatal}$: Số lượng tai nạn chết người bị dự đoán nhầm thành nhẹ/nghiêm trọng (Cần giảm thiểu số này tối đa).

    Bảng Chú thích Ký hiệu (Legend):

    | Ký hiệu | Tên đầy đủ        | Giải thích ý nghĩa |
    |-------|------------------|-------------------|
    | TP    | True Positive    | Dương tính thật: Mô hình dự đoán đúng lớp đang xét (ví dụ: thực tế là **Chết người** và mô hình cũng dự đoán **Chết người**). |
    | TN    | True Negative    | Âm tính thật: Mô hình dự đoán đúng là **không thuộc** lớp đang xét. |
    | FP    | False Positive   | Dương tính giả (báo động giả): Thực tế **không phải** lớp đó nhưng mô hình dự đoán nhầm là có (ví dụ: tai nạn **Nhẹ** nhưng bị dự đoán là **Chết người**). |
    | FN    | False Negative   | Âm tính giả (bỏ sót): Thực tế **thuộc** lớp đó nhưng mô hình không phát hiện ra (ví dụ: tai nạn **Chết người** nhưng bị dự đoán nhầm là **Nhẹ**). |



```python
from sklearn.model_selection import train_test_split

X = df_clean.drop('Accident_Severity', axis=1)
y = df_clean['Accident_Severity']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42,
    stratify=y
)
from sklearn.utils.class_weight import compute_sample_weight

weights = compute_sample_weight(
    class_weight='balanced',
    y=y_train
)
from xgboost import XGBClassifier

# --- 1. KHỞI TẠO MODEL (Sửa ở đây) ---
model = XGBClassifier(
    objective='multi:softprob',
    num_class=3,
    tree_method='hist',
    enable_categorical=True,
    random_state=42,
)

# --- 2. TRAIN MODEL (Sửa ở đây) ---
model.fit(
    X_train,
    y_train,
    sample_weight=weights,
)

```

```python
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import classification_report, confusion_matrix

y_pred = model.predict(X_test)


# Tạo labels cho rõ nghĩa
labels = ['Slight', 'Serious', 'Fatal']

# Tính ma trận nhầm lẫn
cm = confusion_matrix(y_test, y_pred)

# Vẽ Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Dự đoán (Predicted)')
plt.ylabel('Thực tế (Actual)')
plt.title('Confusion Matrix - Hiệu quả dự báo tai nạn')
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.inspection import partial_dependence
import pandas as pd
import numpy as np

# Danh sách biến
features_to_analyze = [
    'Speed_limit',          # Tăng dần (Numeric) -> Line
    'Urban_or_Rural_Area',  # Nominal -> Dot
    'Sex_of_Driver',        # Nominal -> Dot
    'Time_period',          # Tăng dần (Ordinal) -> Line
    'Light_Conditions'      # Nominal -> Dot
]

# Định nghĩa các biến sẽ vẽ đường (còn lại sẽ vẽ Dot)
# Time_period tuy là chữ nhưng có thứ tự tăng dần (Sáng -> Khuya) nên vẽ Line là hợp lý
line_features = ['Speed_limit', 'Time_period']

# Cấu hình màu & Marker
class_info = {
    0: {'name': 'Slight',  'color': '#2ca02c', 'style': '--', 'marker': 'o'}, # Xanh lá
    1: {'name': 'Serious', 'color': '#ff7f0e', 'style': '-.', 'marker': '^'}, # Cam
    2: {'name': 'Fatal',   'color': '#d62728', 'style': '-',  'marker': 's'}  # Đỏ
}

# Setup khung hình
n_cols = 2
n_rows = (len(features_to_analyze) + 1) // n_cols
fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 6 * n_rows))
axes = axes.ravel()

print("Đang vẽ biểu đồ hỗn hợp (Line cho biến tăng dần, Dot cho biến định danh)...")

for idx, feature in enumerate(features_to_analyze):
    ax = axes[idx]

    if feature not in X_train.columns:
        continue

    # --- 1. CHUẨN BỊ DỮ LIỆU ---
    X_plot = X_train.copy()
    if X_plot[feature].isna().any():
        X_plot = X_plot.dropna(subset=[feature])

    # Kiểm tra loại dữ liệu để đưa vào hàm tính toán
    is_categorical = (
        isinstance(X_plot[feature].dtype, pd.CategoricalDtype) or
        X_plot[feature].dtype == 'object' or
        feature == 'Time_period'
    )

    # Time_period là int nhưng cần đối xử như cat để tính toán,
    # nhưng Urban/Sex cần khai báo categorical_features để model hiểu
    cat_feature_param = [feature] if (is_categorical and feature != 'Time_period') else None

    # --- 2. TÍNH TOÁN PDP ---
    try:
        pd_results = partial_dependence(
            model, X_plot, [feature],
            kind='average',
            categorical_features=cat_feature_param
        )

        y_vals = pd_results['average']
        x_vals = pd_results['grid_values'][0] # Dùng key mới của sklearn

        # --- 3. VẼ BIỂU ĐỒ (PHÂN LUỒNG LOGIC) ---

        # === TRƯỜNG HỢP A: VẼ ĐƯỜNG (LINE) CHO BIẾN TĂNG DẦN ===
        if feature in line_features:
            for class_id, info in class_info.items():
                ax.plot(
                    range(len(x_vals)) if is_categorical else x_vals,
                    y_vals[class_id],
                    marker='o', # Vẫn có chấm nhỏ để đánh dấu
                    markersize=6,
                    linewidth=2,
                    linestyle=info['style'],
                    color=info['color'],
                    label=info['name']
                )

        # === TRƯỜNG HỢP B: VẼ CHẤM (DOT) CHO BIẾN NOMINAL ===
        else:
            # Tính toán vị trí lệch (Dodge) để 3 chấm không đè lên nhau
            x_indices = np.arange(len(x_vals))
            # offsets = [-0.15, 0, 0.15] # Lệch Trái - Giữa - Phải

            for class_id, info in class_info.items():
                ax.scatter(
                    x_indices,
                    # + offsets[class_id], # Vị trí + Độ lệch
                    y_vals[class_id],
                    color=info['color'],
                    marker=info['marker'], # Marker khác nhau (Tròn, Tam giác, Vuông)
                    s=100, # Kích thước lớn
                    label=info['name'],
                    zorder=3
                )

                # Vẽ đường nối cực mờ để dẫn mắt (Optional)
                ax.plot(
                    x_indices ,
                    # + offsets[class_id],
                    y_vals[class_id],
                    color=info['color'],
                    alpha=0.2, # Rất mờ
                    linestyle=':'
                )

        # --- 4. XỬ LÝ NHÃN TRỤC HOÀNH ---

        # Nếu là Time_period (Line nhưng cần nhãn chữ)
        if feature == 'Time_period':
            labels = ['Sáng sớm', 'Ban ngày', 'Chiều tối', 'Tối', 'Khuya']
            final_labels = labels if len(x_vals) == len(labels) else x_vals
            ax.set_xticks(range(len(x_vals)))
            ax.set_xticklabels(final_labels, rotation=45, ha='right')

        # Nếu là Nominal Category (Dot)
        elif is_categorical:
            ax.set_xticks(range(len(x_vals)))
            ax.set_xticklabels(x_vals, rotation=45, ha='right')
            # Thêm lưới dọc để phân cách các nhóm
            ax.grid(True, axis='x', linestyle='--', alpha=0.5)

        # Nếu là Numeric (Speed)
        elif feature == 'Speed_limit':
            ax.set_xlabel("Giới hạn tốc độ (mph)")

        # --- 5. TRANG TRÍ ---
        ax.set_title(f"Tác động của '{feature}'", fontsize=12, fontweight='bold')
        ax.set_ylabel("Xác suất dự báo")
        ax.grid(True, axis='y', alpha=0.3)
        ax.legend()

    except Exception as e:
        print(f"❌ Lỗi biến {feature}: {e}")

# Xóa subplot thừa
for j in range(len(features_to_analyze), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.suptitle("PHÂN TÍCH PDP HỖN HỢP\n(Line cho biến có xu hướng - Dot cho biến định danh)",
             y=1.02, fontsize=16, fontweight='bold')
plt.show()
```

  * Ở khu vực tốc độ thấp (20–30 mph), tai nạn xảy ra nhiều nhưng chủ yếu là va chạm nhẹ. Khi tốc độ cao hơn, động năng va chạm lớn, dẫn đến mức độ thương tích nặng hơn. Điều này phản ánh đúng quy luật vật lý và thực tiễn an toàn giao thông.

  * Vì Speed_limit là yếu tố then chốt ảnh hưởng đến mức độ nghiêm trọng, nên cần có các biện pháp hạn chế tốc dộ di chuyển như là giảm tốc độ giới hạn trong khu dân cư, kiểm soát tốc độ nghiêm ngặt trên đường cao tốc

* Biến Urban_or_Rural_Area (Khu vực đô thị / nông thôn)

  * Khu vực nông thôn (Rural) có xác suất tai nạn nghiêm trọng và chết người cao hơn so với đô thị. Khu vực đô thị (Urban) chủ yếu gắn với tai nạn nhẹ.

  * Lí do là vì đường nông thôn thường có ít đèn chiếu sáng,
  tốc độ lưu thông cao,
  khoảng cách đến cơ sở y tế xa. Đô thị có mật độ xe cao nhưng tốc độ thấp → va chạm nhiều nhưng mức độ nhẹ.

  * Vì vậy cần ưu tiên đầu tư an toàn giao thông tại khu vực nông thôn, như là lắp thêm nhiều đèn phát sáng, giảm tốc độ di chuyển ở nơi đông người.

* Biến Sex_of_Driver (Giới tính người lái)


  * Sự khác biệt giữa hai nhóm giới tính không quá lớn, nhưng có xu hướng: nam sẽ có xác xuất xảy ra tai nạn nghiêm trọng hơn nữ

  * Nguyên nhân có thể là do nam là trụ cột gia đình nên sẽ thường lái xe di chuyển nhiều hơn nữ.

  * Vì vậy cần nâng cao nhận thức an toàn giao thông của người dân.
* Biến Time_period (Thời điểm trong ngày)

  * Tai nạn xảy ra vào ban đêm và khuya có xu hướng nghiêm trọng hơn. Ban ngày và sáng sớm có xác suất tai nạn nhẹ cao hơn.

  * Nguyên nhân là do khi lái xe vào ban đêm sẽ có tầm nhìn kém, dễ mệt mỏi và buồn ngủ hơn. Với lưu lượng xe thấp sẽ dẫn đến tốc độ cao, khi gặp tai nạn sẽ nguy hiểm hơn.

  Vì vậy cần tăng cường đèn chiếu sáng, bắt buộc bật đèn xe vào ban đêm, không được lái xe liên tục trong thời gian dài, kiểm soát tốc độ theo khung giờ
* Biến Light_Conditions (Điều kiện ánh sáng)

  * Điều kiện thiếu ánh sáng làm tăng rõ rệt mức độ nghiêm trọng tai nạn. Ánh sáng ban ngày cho kết quả an toàn hơn ánh sáng vào ban đêm.

  * Việc Thiếu ánh sáng làm:

  Giảm khả năng quan sát

  Tăng phản ứng chậm

  Ngay cả khi có đèn đường, mức độ an toàn vẫn thấp hơn ban ngày.


  Vì ánh sáng là yếu tố hạ tầng có thể can thiệp trực tiếp, nên cần có các biện pháp cải thiện, như là nâng cấp độ sáng đèn đường, bổ sung đèn nhiều hơn ở ngã 3,4; tăng cường vật liệu phản quang và biển báo ban đêm; sơn vạch kẻ đường phản quang.
